
# **🌐 1. Scope of the Diagnostic Analysis**

The diagnostic analysis covers the entire macro-socioeconomic, technical, and regulatory ecosystem of electric mobility across all eight (8) East African Community (EAC) Partner States: the Republic of Burundi, the Democratic Republic of the Congo (DRC), the Republic of Kenya, the Republic of Rwanda, the Republic of South Sudan, the Republic of Uganda, the United Republic of Tanzania, and the Federal Republic of Somalia.

-  Temporal Boundary: Baseline evaluation uses historical vehicle registrations, structural trends, and operational tracking parameters spanning 2021 through 2026, with strategic demand, charging network updates, and utility cash generation capacity projected forward under dynamic electrification adoption profiles.
-  Segment Boundary: Micro-level granular telematics focus heavily on high-utilization commercial segments specifically two-wheelers (2W) and three-wheelers (3W) (Boda-Bodas)—while aggregating market fleet dynamics across light commercial vehicles (LCVs), public transport buses, and private personal passenger fleets.

-  Objective 2: Expose Macroeconomic Vulnerabilities. Quantify the national financial exposure of partner states to fossil fuel import dependencies and explicit consumer fuel subsidy strains against the economic stability of transitioning to domestic green power.

-  Objective 3: Conduct a Harmonized Regulatory Audit. Systematically evaluate import tariffs, fiscal incentives, power utility standards, and structural policy variations across all eight states to identify trade anomalies and investment hurdles.

-  Objective 4: Prove Microeconomic Commercial Viability. Model the unit economics, net operator earnings, and Total Cost of Ownership (TCO) break-even horizons for active e-mobility deployments compared directly to legacy internal combustion alternatives.

-  Objective 5: Determine Grid and Circular Value Chain Readiness. Diagnose national power grid capacities, clean generation asset profiles, spatial charging layout configurations, and industrial battery end-of-life framework indicators.

## **⚙️ 4. Data Mining, Wrangling, & Databasing Methodology**

**Phase B: Cleaning & Anomaly Correction**

Using Python (pandas, numpy), data cleaning resolved critical telemetry anomalies:

**Phase C: Merging & Relational Integration**

• The countries dimension table acts as the central hub.

• Rider telematics tables (11 and 13) were aggregated by country-day averages before merging with the main dashboard layer. This keeps the final dashboard fast and easy to navigate for non-technical stakeholders.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --- 1. Coerce projected column to numeric (turns "Not Found" into NaN) ---
main['Projected rate of motorization (2050)'] = pd.to_numeric(
    main['Projected rate of motorization (2050)'], errors='coerce'
)
main['Current Rate of Motorization (2016)'] = pd.to_numeric(
    main['Current Rate of Motorization (2016)'], errors='coerce'
)

# --- 2. Category functions ---
def cat_current(v):
    if pd.isna(v): return 'Unknown'
    if v < 10:   return 'Low'
    if v <= 20:  return 'Medium'
    return 'High'

def cat_projected(v):
    if pd.isna(v): return 'Unknown'
    if v < 50:    return 'Low'
    if v <= 80:   return 'Medium'
    if v <= 100:  return 'High'
    return 'Very High'

df = main.copy()
df['Curr_Cat'] = df['Current Rate of Motorization (2016)'].apply(cat_current)
df['Proj_Cat'] = df['Projected rate of motorization (2050)'].apply(cat_projected)

# Sort by projected (NaN last)
df = df.sort_values('Projected rate of motorization (2050)', ascending=True, na_position='first')

# --- 3. Plot ---
y = np.arange(len(df))
h = 0.38

fig, ax = plt.subplots(figsize=(13, 7), dpi=110)
ax.set_facecolor('#F8FAFC')

bars_p = ax.barh(y + h/2, df['Projected rate of motorization (2050)'], h,
                 color='#0284C7', edgecolor='#0F172A', linewidth=1,
                 label='Projected (2050)')
bars_c = ax.barh(y - h/2, df['Current Rate of Motorization (2016)'], h,
                 color='#94A3B8', edgecolor='#0F172A', linewidth=1,
                 label='Current (2016)')

# Value + category labels
for bar, val, cat in zip(bars_p, df['Projected rate of motorization (2050)'], df['Proj_Cat']):
    label = f'{val:.0f}  ({cat})' if pd.notna(val) else 'No data'
    ax.text(bar.get_width() + 2,
            bar.get_y() + bar.get_height()/2,
            label, va='center', ha='left',
            color='#0369A1', weight='bold', size=9)

for bar, val, cat in zip(bars_c, df['Current Rate of Motorization (2016)'], df['Curr_Cat']):
    label = f'{val:.1f}  ({cat})' if pd.notna(val) else 'No data'
    ax.text(bar.get_width() + 2,
            bar.get_y() + bar.get_height()/2,
            label, va='center', ha='left',
            color='#475569', size=9)

# Threshold lines
for x in (10, 20):
    ax.axvline(x, color='#94A3B8', linestyle=':', linewidth=1.2)
for x in (50, 80, 100):
    ax.axvline(x, color='#0284C7', linestyle='--', linewidth=1.2, alpha=0.7)

ax.set_yticks(y)
ax.set_yticklabels(df['Country'], fontsize=11, weight='bold', color='#1E293B')
ax.set_xlabel('Motorization Rate (vehicles per 1,000 inhabitants)',
              fontsize=11, weight='semibold', color='#334155', labelpad=12)
ax.set_title('Current  vs Projected (2050) Motorization Rates by Country',
             fontsize=14, weight='bold', color='#1E293B', pad=20)

for s in ['top', 'right']:
    ax.spines[s].set_visible(False)
ax.grid(axis='x', linestyle='--', color='#E2E8F0', alpha=0.6)
ax.legend(loc='lower right', frameon=True, facecolor='white', edgecolor='#E2E8F0')

plt.tight_layout()
plt.show()
```

Analyzing Fiscal Tariffs—specifically comparing the import duties, excise taxes, value-added taxes (VAT), and registration fees applied to Internal Combustion Engine (ICE) vehicles versus Electric Vehicles (EVs)—reaches the absolute core of your diagnostic study.
This dataset underpins Chapter 4.0 (Policy Development Process), Chapter 8.0 (Incentives), and Chapter 9.0 (Importation of EVs). Before you run any code or look at the rows, you must frame this data around its critical structural impacts on the region:

1. It Quantifies the "Upfront Price-Parity Gap" (The Consumer Hurdle)

As we established in your Group A income metrics, individual consumer purchasing power across the EAC hovers within a tight baseline tier ($1,000–$2,000 per capita in the strongest core markets). Because electric vehicles carry an intense manufacturing cost premium due to their advanced lithium battery chemistries, they are naturally more expensive upfront than a legacy gasoline equivalent.

The Impact: Tariffs are the primary tool a government has to artificially force price parity. This dataset calculates the Tariff Advantage Delta across borders. If the total tax burden on an EV is identical to or higher than an ICE car, the technology is dead on arrival for local buyers. A significant, negative tax delta proves mathematically that a country's trade policy is actively strangling its transport transition.

2. It Exposes "Hidden Component Penalties" in the Value Chain (Chapter 10.0)

Many early-stage customs structures carry a major design flaw: they introduce zero-rated taxes on a fully assembled, imported electric car but continue to heavily tax imported lithium batteries, electric drivetrains, and charging hardware components under standard industrial raw material or luxury electronic brackets.

The Impact: This creates a severe inversion that penalizes local value chains. It makes it cheaper to import finished foreign vehicles than to build local assembly, battery-swapping, and component-manufacturing plants within the EAC, contradicting the industrial growth goals of the EAC Treaty.

3. It Highlights Regional Customs Friction and Smuggling Arbitrage

The eight EAC partner states share a theoretical Common External Tariff (CET), but national deviations, local tax exemptions, and different VAT applications create steep tariff borders between contiguous neighbors.

The Impact: If Kenya offers a 10% total tax burden on electric motorcycles while neighboring Uganda or Tanzania applies a 45% standard tariff, it creates a massive price distortion. This encourages illegal cross-border smuggling arbitrage, breaks the regional trading bloc's unity, and prevents manufacturing networks from setting up optimized, regional logistics paths.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Use your existing variable 'fiscal' directly
# 1. Clean and convert columns to floats to protect against mixed text/float type mismatches
for col in ['Import Duty (ICE)', 'Import Duty (EV)', 'Excise Duty (ICE)', 'Excise Duty (EV)', 'VAT Rate (ICE)', 'VAT Rate (EV)']:
    # FIX: Safely parse the column directly from the 'fiscal' dataframe
    fiscal[col] = pd.to_numeric(fiscal[col], errors='coerce').fillna(0)
    
    # If the column values are stored as whole numbers (e.g. 16 for 16%), convert to decimals
    if fiscal[col].max() > 1.0:
        fiscal[col] = fiscal[col] / 100.0

# 2. Compute the True Cascading Cumulative Port Tax Multiplier mathematically
# Formula accounts for standard compounding customs architecture: (1 + Import) * (1 + Excise) * (1 + VAT) - 1
fiscal['Total_ICE_Tax_Pct'] = ((1 + fiscal['Import Duty (ICE)']) * (1 + fiscal['Excise Duty (ICE)']) * (1 + fiscal['VAT Rate (ICE)']) - 1) * 100
fiscal['Total_EV_Tax_Pct'] = ((1 + fiscal['Import Duty (EV)']) * (1 + fiscal['Excise Duty (EV)']) * (1 + fiscal['VAT Rate (EV)']) - 1) * 100

# Calculate the precise Tariff Advantage Delta (Positive means EV is taxed less)
fiscal['Tariff_Advantage_Delta'] = fiscal['Total_ICE_Tax_Pct'] - fiscal['Total_EV_Tax_Pct']
fiscal = fiscal.sort_values(by='Tariff_Advantage_Delta', ascending=False)

# 3. Canvas initialization and professional design
fig, ax = plt.subplots(figsize=(12, 7), dpi=100)
ax.set_facecolor('#F8FAFC')

y_indices = np.arange(len(fiscal))
bar_height = 0.35

# 4. Plot True Cumulative Compounded Border Taxes side-by-side
bars_ice = ax.barh(y_indices - bar_height/2, fiscal['Total_ICE_Tax_Pct'], bar_height, 
                    label='Total Compounded Tax Burden: ICE', color='#64748B', edgecolor='#0F172A', linewidth=1)

bars_ev = ax.barh(y_indices + bar_height/2, fiscal['Total_EV_Tax_Pct'], bar_height, 
                   label='Total Compounded Tax Burden: EV', color='#10B981', edgecolor='#0F172A', linewidth=1)

# 5. Add exact compounded values onto the tips of the bars for non-tech clarity
for bar in bars_ice:
    width = bar.get_width()
    if width > 0:
        ax.text(width + 1.0, bar.get_y() + bar.get_height()/2, f"{width:.1f}%", 
                va='center', ha='left', color='#475569', size=9)

for bar in bars_ev:
    width = bar.get_width()
    if width > 0:
        ax.text(width + 1.0, bar.get_y() + bar.get_height()/2, f"{width:.1f}%", 
                va='center', ha='left', color='#065F46', weight='bold', size=10)

# Professional typography and label clean-up using your full country name map
ax.set_title('EAC Port Custom Diagnostics: Total Cumulative Vehicle Tax Burdens\nQuantifying Compounded Port Duties, Excise Taxes, and VAT Applications (ICE vs. Electric Vehicles)', 
             fontsize=13, pad=25, weight='bold', color='#1E293B')
ax.set_xlabel('True Cumulative Compounded Port Tax Burden (%)', fontsize=11, labelpad=12, weight='semibold', color='#334155')
ax.set_yticks(y_indices)

# Map the exact country text cleanly
iso_to_name = {
    'KE': 'Kenya', 'TZ': 'Tanzania', 'UG': 'Uganda', 'RW': 'Rwanda', 
    'BI': 'Burundi', 'SS': 'South Sudan', 'SO': 'Somalia', 'CD': 'DR Congo',
    'Kenya': 'Kenya', 'Tanzania': 'Tanzania', 'Uganda': 'Uganda', 'Rwanda': 'Rwanda',
    'Burundi': 'Burundi', 'South Sudan': 'South Sudan', 'Somalia': 'Somalia', 'DR Congo': 'DR Congo'
}
full_country_names = [iso_to_name.get(c, c) for c in fiscal['Country']]
ax.set_yticklabels(full_country_names, fontsize=11, weight='bold', color='#1E293B')

# Expand horizontal limit safely to leave plenty of room for value tags
ax.set_xlim(0, max(fiscal['Total_ICE_Tax_Pct'].max(), fiscal['Total_EV_Tax_Pct'].max()) + 15)

for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)
ax.grid(axis='x', linestyle='--', color='#E2E8F0', alpha=0.5, zorder=1)
ax.legend(loc='upper right', frameon=True, facecolor='white', edgecolor='#E2E8F0')

plt.tight_layout()
plt.show()

```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Assuming your variable 'eri' is already loaded in your environment.
# We create a temporary working copy to safely clean and process the data.
df_plot = eri.copy()

# 1. Convert 'Not Found' to numeric NaN for proper sorting and plotting
df_plot["2024 ERI Score Num"] = pd.to_numeric(
    df_plot["2024 ERI Score"].replace("Not Found", np.nan), errors="coerce"
)

# Sort the dataset so the highest scores are at the top, and 'Not Found' countries appear at the bottom
df_plot = df_plot.sort_values(
    by="2024 ERI Score Num", ascending=False, na_position="last"
).reset_index(drop=True)

# 2. Build the corporate visualization canvas
sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(15, 8))

# Map discrete palettes including a distinct neutral gray for 'Not Found' status
status_colors = {
    "High": "#1b4d3e",  # Deep Emerald
    "Substantial": "#2e8b57",  # Sea Green
    "Moderate": "#e6a100",  # Amber
    "Low": "#c0392b",  # Dark Red
    "Not Found": "#95a5a6",  # Neutral Slate Gray
}

# Fill NaN statuses with 'Not Found' string to align with our color map legend
df_plot["Score Rank Status"] = df_plot["Score Rank Status"].replace(
    "Not Found", "Not Listed/Developing"
)
status_colors["Not Listed/Developing"] = "#95a5a6"

# Create a placeholder column for plotting widths (give 'Not Found' a minor baseline indicator)
df_plot["Plot_Width"] = df_plot["2024 ERI Score Num"].fillna(0.02)

# 3. Generate the horizontal bar plot
barplot = sns.barplot(
    x="Plot_Width",
    y="EAC Country",
    hue="Score Rank Status",
    palette=status_colors,
    data=df_plot,
    dodge=False,
    ax=ax,
    edgecolor="black",
    linewidth=0.8,
)

# 4. Annotate Contextual Diagnostics on top of the bars
for i, row in df_plot.iterrows():
    is_not_found = pd.isna(row["2024 ERI Score Num"])

    if is_not_found:
        # Label formatting for missing/developing frameworks
        score_text = "Data Not Listed / Developing"
        ax.text(
            0.03,
            i,
            f"{score_text} — {row['Key Insights & Regulatory Status']}",
            va="center",
            ha="left",
            fontweight="bold",
            fontsize=10.5,
            color="#7f8c8d",
            style="italic",
        )
    else:
        # Label formatting for active scores
        score_text = f"{row['2024 ERI Score Num']:.3f}"
        ax.text(
            row["2024 ERI Score Num"] + 0.01,
            i,
            score_text,
            va="center",
            ha="left",
            fontweight="bold",
            fontsize=11,
            color="#2c3e50",
        )

        # Inline diagnostic annotation for established frameworks
        rank_text = f" [Continental Rank: #{int(row['Continental Rank'])}] — {row['Key Insights & Regulatory Status']}"
        ax.text(
            0.02,
            i,
            rank_text,
            va="center",
            ha="left",
            fontsize=10.5,
            color="white" if row["2024 ERI Score Num"] > 0.5 else "black",
            fontweight="medium",
        )

# 5. Framing and Layout Refinements
ax.set_title(
    "Regulatory Readiness Evaluation\n"
    "Based on the African Development Bank 2024 Electricity Regulatory Index (ERI)",
    fontsize=15,
    fontweight="bold",
    pad=25,
    color="#1a252f",
)
ax.set_xlabel(
    "2024 ERI Score Metric (Countries with unestablished independent regulators set to baseline tracking)",
    fontsize=11,
    labelpad=12,
)
ax.set_ylabel("EAC Partner State", fontsize=12, labelpad=12)
ax.set_xlim(0, 1.1)

# Format explicit structural constraints onto the graph
ax.legend(
    title="Regulatory Status Classification",
    title_fontsize="11",
    loc="lower right",
    frameon=True,
    facecolor="white",
    edgecolor="#bdc3c7",
)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Sort on clean energy percentage for proper scale progression
egc_sorted_score = egc.sort_values(by="% Clean", ascending=True)

plt.figure(figsize=(11, 5.5))

# Define clean color palette strings
color_low = "#ee5253"     # Red
color_medium = "#ff9f43"  # Orange
color_high = "#10ac84"    # Green

# Dynamically build color and label vectors directly from the percentages
hue_colors = []
tier_labels = []

for pct in egc_sorted_score["% Clean"].values:
    if pct < 33.0:
        hue_colors.append(color_low)
        tier_labels.append("Low")
    elif pct <= 66.0:
        hue_colors.append(color_medium)
        tier_labels.append("Medium")
    else:
        hue_colors.append(color_high)
        tier_labels.append("High")

# Build the horizontal scorecard graph
bars = plt.barh(egc_sorted_score["Country"], egc_sorted_score["% Clean"], color=hue_colors, edgecolor="black", height=0.6)

# Overlay policy thresholds on the graph layout
plt.axvline(x=33.0, color=color_low, linestyle=":", alpha=0.8, linewidth=2, label="Low Threshold (<33%)")
plt.axvline(x=66.0, color=color_medium, linestyle=":", alpha=0.8, linewidth=2, label="High Threshold (>66%)")

plt.title("EAC E-Mobility Strategic Decarbonization Readiness Scoring",fontweight='bold')
plt.xlabel("Clean Power Generation Proportional Base (%)")
plt.ylabel("Country")

# Right-aligned text anchors using the freshly generated lists
for bar, label, color in zip(bars, tier_labels, hue_colors):
    width = bar.get_width()
    plt.text(width + 1.5, bar.get_y() + bar.get_height()/2, f"Tier: {label}", va="center", weight="bold", color=color)

plt.xlim(0, 115)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

## **Daily Operations**

Analyzing Daily Operations represents the absolute baseline of micro-level diagnostics. While trip-level data captures individual route efficiencies, the daily operational scale reflects the actual economic reality of a Boda-Boda rider’s life. By evaluating data aggregated at the daily operational level, we can isolate the daily fixed costs from the highly fluid variable costs, unlocking distinct diagnostic insights for both ICE and EV platforms:

- For ICE Motorbikes (The Volatile OpEx Trap): Daily operations expose the harsh reality of the "fuel hand-to-mouth" cycle.
```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# 1. Clean and convert operating hours to numerical daily values if they are strings (e.g., "24 Hours" or "08:00-17:00")
def parse_operating_hours(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).lower().strip()
    if '24' in val_str or 'always' in val_str or 'non-stop' in val_str:
        return 24.0
    # Handle ranges like "08:00-17:00" or "8am-5pm"
    try:
        if '-' in val_str:
            parts = val_str.split('-')
            # Extract basic numeric hour values
            start = float(''.join(c for c in parts[0] if c.isdigit() or c == '.'))
            end = float(''.join(c for c in parts[1] if c.isdigit() or c == '.'))
            # Rough normalization for standard am/pm or 24h formats
            if 'pm' in parts[1] and end < 12: end += 12
            if 'pm' in parts[0] and start < 12: start += 12
            duration = end - start
            return duration if duration > 0 else duration + 24
    except:
        pass
    # Fallback default if numeric parsing fails but valid string exists
    return 9.0

# Create a copy and map numerical operational windows
charging_hours = ev_charge.copy()
if charging_hours['operating_hours'].dtype == 'O':
    charging_hours['daily_hours_open'] = charging_hours['operating_hours'].apply(parse_operating_hours)
else:
    charging_hours['daily_hours_open'] = charging_hours['operating_hours']

# 2. Identify the top 4 infrastructure providers across the whole regional dataset
top_4_providers = charging_hours['provider'].value_counts().nlargest(4).index
filtered_infra = charging_hours[charging_hours['provider'].isin(top_4_providers)]

# 3. Plot the categorical distribution using a clean, easily interpretable factorplot/stripplot
plt.figure(figsize=(13, 7))
sns.stripplot(
    data=filtered_infra,
    x='country',
    y='daily_hours_open',
    hue='provider',
    dodge=True,         # Separates the providers into clear, non-overlapping columns
    jitter=0.2,         # Spreads out identical values for visual clarity
    alpha=0.75,
    size=9,
    palette='Set1'
)

plt.title('Operational Availability Envelopes: Top 4 Providers vs. Country and Daily Operating Hours', fontsize=13, fontweight='bold', pad=15)
plt.xlabel('EAC Member State', fontsize=11, fontweight='bold')
plt.ylabel('Daily Operational Availability Window (Hours Open per Day)', fontsize=11, fontweight='bold')
plt.yticks(range(0, 26, 4))
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(title='Top 4 Infrastructure Operators', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

**?? Core Analytical Observations from the Output**

1. The highest-performing providers maintain wider daily operational windows, while weaker operators are more constrained by shorter service hours and lower accessibility.

2. The graph confirms that the real competitive advantage is not only station count but the duration and reliability of service availability across the day.

3. In practical terms, a provider with broader daily coverage can support more productive fleet operations than a provider with a large but narrow service window.
