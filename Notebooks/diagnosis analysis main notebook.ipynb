
# **🌐 1. Scope of the Diagnostic Analysis**

The diagnostic analysis covers the entire macro-socioeconomic, technical, and regulatory ecosystem of electric mobility across all eight (8) East African Community (EAC) Partner States: the Republic of Burundi, the Democratic Republic of the Congo (DRC), the Republic of Kenya, the Republic of Rwanda, the Republic of South Sudan, the Republic of Uganda, the United Republic of Tanzania, and the Federal Republic of Somalia.

-  Temporal Boundary: Baseline evaluation uses historical vehicle registrations, structural trends, and operational tracking parameters spanning 2021 through 2026, with strategic demand, charging network updates, and utility cash generation capacity projected forward under dynamic electrification adoption profiles.
-  Segment Boundary: Micro-level granular telematics focus heavily on high-utilization commercial segments specifically two-wheelers (2W) and three-wheelers (3W) (Boda-Bodas) while aggregating market fleet dynamics across light commercial vehicles (LCVs), public transport buses, and private personal passenger fleets.


## **🎯 2. Objectives of the Diagnostic Analysis**

-  Objective 1: Establish the Quantitative Baseline; Define the exact scale of the active vehicle fleet, current electrification rates, and tracking metrics across the region to calibrate mid-term adoption trajectories.

-  Objective 2: Expose Macroeconomic Vulnerabilities; Quantify the national financial exposure of partner states to fossil fuel import dependencies and explicit consumer fuel subsidy strains against the economic stability of transitioning to domestic green power.

-  Objective 3: Conduct a Harmonized Regulatory Audit; Systematically evaluate import tariffs, fiscal incentives, power utility standards, and structural policy variations across all eight states to identify trade anomalies and investment hurdles.

-  Objective 4: Prove Microeconomic Commercial Viability; Model the unit economics, net operator earnings, and Total Cost of Ownership (TCO) break-even horizons for active e-mobility deployments compared directly to legacy internal combustion alternatives.

-  Objective 5: Determine Grid and Circular Value Chain Readiness; Diagnose national power grid capacities, clean generation asset profiles, spatial charging layout configurations, and industrial battery end-of-life framework indicators.


## **🗃️ 3. Data Governance & Dataset Understanding**

The structural assessment relies on a relational database architecture consisting of 31 specialized datasets. 

(Refer to your master data checklist for the exhaustive technical details of all 31 tables, structured around the foundational core below):

- Foundational Spine: 1. countries acts as the primary dimensional anchor table holding the entire relational ecosystem together via standardized country codes.
- Empirical Baseline Engines: 2. market_fleet, 3. policy_inventory, 4. fiscal_tariffs, and 5. power_system map structural state readiness profiles.
- Micro-Level Telematics Repositories: 11. Fuel_motorcycle_daily_operations through 14. Electric_motorcycle_trip_data drive high-resolution rider operational modeling.



## **⚙️ 4. Data Mining, Wrangling, & Databasing Methodology**

To transition raw, disparate data inputs into an integrated analysis platform, a multi-stage Data Wrangling Pipeline was built using Python.


**Phase A: Databasing & Structuring**

Raw unstructured tracking logs, telematics metrics, and country policy updates were collected virtually from ministerial platforms, telematics operators, and utility entities. Data was structured within a relational format using ISO-3166-1 Alpha-3 country codes (KEN, UGA, TZA, etc.) as the standardized global foreign key to eliminate text string naming mismatches.

**Phase B: Cleaning & Anomaly Correction**

Using Python (pandas, numpy), data cleaning resolved critical telemetry anomalies:

• Missing Values: Qualitative profiles were used to fill data gaps for data-scarce environments without skewing quantitative fleet averages.


**Phase C: Merging & Relational Integration**

The tables were organized into a Star Schema Design:

• The countries dimension table acts as the central hub.

• Fleet data, tariff configurations, power parameters, and macro metrics serve as surrounding factual indicators.

• Rider telematics tables (11 and 13) were aggregated by country-day averages before merging with the main dashboard layer. This keeps the final dashboard fast and easy to navigate for non-technical stakeholders.

# **GROUP A**

## **Countries**

This dataset  serves as the fundamental anchor and critical foundation for the entire EAC E-mobility Diagnosis Study.

Before mapping out vehicle fleets, grid capacities, or charging infrastructures, it is necessary to establish the underlying macroeconomic constraints of the region. Group A provides the structural boundaries for demographic scale, economic output, purchasing power, and institutional timelines across all eight East African Community (EAC) partner states: Burundi, the Democratic Republic of the Congo (DRC), Kenya, Rwanda, Somalia, South Sudan, Tanzania, and Uganda.

**Why Group A is the Foundation of the Study**

1. Defining Total Addressable Market (TAM): Population metrics establish the human scale and urban density boundaries that dictate the potential volume of commuting populations and logistics corridors.

2. Quantifying Affordability & Purchasing Power: GDP per capita (Nominal) acts as a strict proxy for consumer purchasing power. It determines where high-upfront-cost electric vehicles (EVs) are economically viable vs. where low-cost asset financing models (such as battery-swapping for two-wheelers) are mandatory.

3. Macro-Fiscal Scale: National GDP figures specify the fiscal capacity of partner governments to absorb tax incentives (e.g., zero-rating VAT, subsidizing EV tariffs) or secure sovereign infrastructure investments.

4. Institutional Readiness Windows: Accession years and member statuses identify the legal maturity of each state within the EAC treaty framework, highlighting which markets can move uniformly under regional e-mobility standards vs. those operating under transitional bilateral rules.

```python
countries = pd.read_csv('eac_emobility_baseline_data.csv')
countries
```

```python
countries.columns
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Setup professional visual styling
sns.set_theme(style="whitegrid")
plt.rcParams["font.family"] = "sans-serif"
countries_sorted = countries.sort_values(by="gdp_usd", ascending=False)

fig, ax1 = plt.subplots(figsize=(12, 6))
ax2 = ax1.twinx()

# Plot Absolute GDP (Bars)
sns.barplot(
    x="country_name",
    y=countries_sorted["gdp_usd"] / 1e9,
    data=countries_sorted,
    alpha=0.7,
    color="#2c3e50",
    ax=ax1,
    label="Nominal GDP (Left Axis)",
)
ax1.set_ylabel("Nominal GDP (Billion USD)", color="#2c3e50", weight="bold")
ax1.tick_params(axis="y", labelcolor="#2c3e50")
ax1.set_xticklabels(ax1.get_xticklabels(), rotation=15, ha="right")

# Plot GDP Per Capita (Line)
sns.lineplot(
    x="country_name",
    y="gdp_per_capita",
    data=countries_sorted,
    color="#e74c3c",
    marker="o",
    linewidth=3,
    markersize=8,
    ax=ax2,
    label="GDP Per Capita (Right Axis)",
)
ax2.set_ylabel("GDP Per Capita (USD)", color="#e74c3c", weight="bold")
ax2.tick_params(axis="y", labelcolor="#e74c3c")
ax2.grid(False)

plt.title(
    "EAC Strategic Matrix: Absolute GDP Scale vs. Individual Purchasing Power",
    fontsize=14,
    weight="bold",
    pad=15,
)
ax1.set_xlabel("Partner State", fontsize=11, labelpad=10)

# Unify Legend
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper right")

fig.tight_layout()
plt.show()
```

**🔍 Core Analytical Observations from the Output**

1. The Hegemonic Market Leader (Kenya):

The strategic matrix establishes Kenya as the absolute economic anchor of the EAC region, dominating on both macro and micro scales. Kenya holds the largest Nominal GDP at 147 Billion USD paired with the highest individual purchasing power, leading the bloc with a GDP Per Capita of 2,550 USD. This dual-dominance confirms why it serves as the primary gateway for regional commercial capital.

2. The High-Volume, Low-Purchasing Paradox (DR Congo):

A stark structural mismatch is visible in the DR Congo, which commands the second-largest economic scale in the bloc with a Nominal GDP of 123 Billion USD. However, its individual purchasing power sharply decouples, crashing to a GDP Per Capita of 1,100 USD. This highlights an economy driven by massive localized asset extraction or high aggregate population volume, rather than widely distributed consumer purchasing power.

3. The Stable Intermediate Tier (Tanzania & Uganda):

Tanzania and Uganda form a solid, mid-tier economic corridor. Tanzania demonstrates robust positioning with a Nominal GDP of 95 Billion USD and a rebounding GDP Per Capita of 1,350 USD (the second-highest in the dataset). Uganda tracks closely with a Nominal GDP of 63 Billion USD and a steady GDP Per Capita of 1,250 USD, showing highly aligned economic density profiles.

4. The Micro-Scale and Frontier Floors (Rwanda down to Burundi):

The right half of the matrix reveals severe scale constraints across the remaining partner states. While Rwanda maintains a stable consumer profile (1,150 USD GDP Per Capita), its absolute market size is highly constrained at a Nominal GDP of 16 Billion USD. Moving further down the frontier, Somalia, South Sudan, and Burundi face compounding structural distress, culminating in Burundi marking the absolute baseline floor of the region with an economy of just 3 Billion USD Nominal GDP and a nominal GDP Per Capita of 200 USD.

**💡 Strategic Policy Insights for the E-Mobility Diagnosis Study**

1. Aligning Capital Expenditure with Consumer Purchasing Realities (Chapter 18.0):

The dramatic variance in GDP Per Capita from Kenya ($2,550) down to Burundi ($200) proves that a single, unified vehicle asset model cannot be deployed across the EAC. In high-purchasing power markets like Kenya, policy frameworks should encourage direct consumer financing models for premium electric vehicles. Conversely, in low per-capita zones like the DR Congo, Tanzania, and Uganda, policies must favor heavily subsidized Battery-as-a-Service (BaaS) operational models to eliminate the high upfront component costs that consumer wallets cannot absorb.

2. Leveraging Absolute Scale for Centralized Manufacturing Hubs (Chapter 1.0):

The sheer volume of Nominal GDP concentrated in the Top 3 markets Kenya, DR Congo, and Tanzania indicates where regional assembly and battery manufacturing plants must be anchored. Our diagnostic study should recommend an EAC Industrial Localization Strategy, advising international e-mobility OEMs to concentrate heavy manufacturing and component supply chains inside these high-scale economies to maximize domestic market utilization, while treating smaller economies like Rwanda and Burundi as light retail and distribution endpoints.

3. Tailoring Micro-Grant Injections for Frontier Grid Survivability (Chapter 7.0):

For frontier states like Somalia, South Sudan, and Burundi, where both nominal scale and individual purchasing power collapse to near-zero baselines, commercial venture capital will not naturally flow. To kickstart e-mobility in these zones, the policy framework must bypass traditional market dynamics entirely. Ministries of Energy must rely on Concessional DFI Micro-Grants to fund state-backed, solar-powered e-mobility micro-grids, ensuring that clean transport access is deployed as a public welfare service rather than a profit-driven commercial enterprise.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 7))

# Generate scatter bubble plot
sns.scatterplot(
    x=countries["gdp_usd"] / 1e9,
    y="gdp_per_capita",
    size=countries["population"] / 1e6,
    hue="notes",
    sizes=(200, 1500),
    alpha=0.8,
    palette="deep",
    data=countries,
)

# Annotate country codes next to bubbles
for i, txt in enumerate(countries["country_name"]):
    plt.annotate(
        txt,
        (countries["gdp_usd"].iloc[i] / 1e9 + 2, countries["gdp_per_capita"].iloc[i]),
        fontsize=10,
        weight="bold",
    )

# Add strategic diagnostic quadrants
plt.axhline(
    y=1100, color="#7f8c8d", linestyle="--", alpha=0.6, label="Affordability Tier Limit"
)
plt.axvline(
    x=50, color="#7f8c8d", linestyle="--", alpha=0.6, label="Macro Market Scale Limit"
)

plt.title(
    "EAC Market Positioning: Affordability vs. Economic Scale\n(Bubble size relative to Population in Millions)",
    fontsize=13,
    weight="bold",
    pad=15,
)
plt.xlabel("Nominal GDP (Billion USD)", fontsize=11)
plt.ylabel("GDP Per Capita (USD)", fontsize=11)
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left", title="Status / Metrics")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

**🔍 Core Analytical Observations from the Output**

1. The Outlier Frontier (Kenya):

The bubble chart establishes Kenya as an undisputed outlier, sitting entirely alone in the top-right quadrant of the matrix. It represents a highly mature consumer profile with a GDP Per Capita of 2,500 USD and a dominant Nominal GDP of 147 Billion USD, while maintaining a robust, mid-sized demographic anchor (55 million population). This positioning places Kenya well above both the Affordability Tier Limit and the Macro Market Scale Limit.

2. The High-Density Mid-Tier (Tanzania & Uganda):

Tanzania and Uganda cluster closely together as stable, high-potential markets bridging the macro limits. Uganda follows with 1,200 USD GDP Per Capita at a 64 Billion USD Nominal GDP scale. Both countries hold strategic advantages due to their founder-member status (blue bubbles) and highly balanced economic-to-demographic proportions.

3. The High-Volume Demographic Trap (DR Congo):

The DR Congo creates a striking visual contrast, represented by the largest bubble on the chart (100 million population). It breaks far past the Macro Market Scale Limit with a Nominal GDP of 123 Billion USD, yet its individual purchasing power sags completely below the line, sitting at a GDP Per Capita of 1,050 USD. This highlights a massive aggregate market where extreme consumer volume masks low individual purchasing capacity.

4. The Truncated and Fragile Baselines (Rwanda down to Burundi):

The bottom-left quadrant contains a mix of constrained or fragile economies failing to cross either threshold line.
```python
import pandas as pd


def clean_segment_value(val):
    if pd.isna(val) or isinstance(val, (int, float)):
        return val

    # Convert to string and clean up spacing/case
    val_str = str(val).strip().lower().replace("$", "")

    try:
        # Handle "million" text values
        if "million" in val_str:
            num_part = val_str.replace("million", "").strip()
            return float(num_part) * 1e6
        # Handle "billion" text values
        elif "billion" in val_str:
            num_part = val_str.replace("billion", "").strip()
            return float(num_part) * 1e1
        # Fallback for standard numeric strings
        return float(val_str)
    except ValueError:
        return 0.0


# Clean the column in place before plotting
market["est_segment_value_usd"] = market["est_segment_value_usd"].apply(
    clean_segment_value
)
print("[SUCCESS] Column converted to clean numbers ready for plotting!")
```

```python
import pandas as pd

# 1. Create a copy to protect your original dataset
market_clean = market.copy()

# 2. Make sure values are numeric (handling the conversion to millions)
if market_clean["est_segment_value_usd"].dtype == "O":

    def convert_to_num(val):
        if pd.isna(val):
            return 0.0
        val_str = str(val).strip().lower().replace("$", "")
        if "million" in val_str:
            return float(val_str.replace("million", "").strip()) * 1e6
        if "billion" in val_str:
            return float(val_str.replace("billion", "").strip()) * 1e9
        return float(val_str) if val_str else 0.0

    market_clean["est_segment_value_usd"] = market_clean[
        "est_segment_value_usd"
    ].apply(convert_to_num)

# Create the working value column in millions
market_clean["value_millions"] = market_clean["est_segment_value_usd"] / 1e6

# 3. Isolate rows that are NOT already combined
split_countries = market_clean[
    market_clean["vehicle_segment"] != "All Vehicles (Combined)"
]

# 4. Mathematically sum them up grouped by country and year
calculated_totals = (
    split_countries.groupby(["country_iso", "year"])
    .agg(
        {
            "value_millions": "sum",
            "est_segment_value_usd": "sum",
            "data_source": "first",
            "data_confidence": "first",
            "notes": "first",
        }
    )
    .reset_index()
)

# 5. Label these new rows exactly as 'All Vehicles (Combined)'
calculated_totals["vehicle_segment"] = "All Vehicles (Combined)"
calculated_totals["propulsion_type"] = "Combined"

# 6. Append the calculated totals back into the main plotting dataframe
plot_ready_df = pd.concat([market_clean, calculated_totals], ignore_index=True)

print(
    "[SUCCESS] Totals calculated and aligned. Ready to generate the multi-bar plot!"
)




```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(13, 6))

# Map Confidence Tiers to numeric point sizes for the scatter layout
confidence_sizes = {"High": 350, "Medium-High": 220, "Medium": 120, "Low-Medium": 60, "Low": 25}
market["marker_size"] = market["data_confidence"].map(confidence_sizes).fillna(50)

# Generate categorical scatter plot across countries
sns.scatterplot(
    x="country_iso",
    y="import_or_sales_count",
    hue="propulsion_type",
    size="marker_size",
    sizes=(30, 400),
    alpha=0.85,
    palette=["#2ecc71", "#c0392b"],  # Green for EV, Red for ICE
    edgecolor="black",
    linewidth=1.2,
    data=market
)

# Apply log axis to handle the extreme scale difference between EV flows and massive ICE flows cleanly
plt.yscale("log")

plt.title(
    "Annual Technology Velocity & Registration Registry Integrity by Country\n"
    "(Marker size reflects Data Confidence Tier)",
    fontsize=14,
    weight="bold",
    pad=15,
)
plt.xlabel("Partner State (ISO Code)", fontsize=11, labelpad=10)
plt.ylabel("Annual Imports / Sales Count ", fontsize=11)

# Clean up custom legend layout to prevent overlapping sizing values
handles, labels = plt.gca().get_legend_handles_labels()
# Keep only propulsion types in the legend to stay punchy and readable
plt.legend(handles[0:3], labels[0:3], title="Propulsion Type", loc="lower left", frameon=True)

plt.grid(True, linestyle=":", alpha=0.5, which="both")
plt.tight_layout()
plt.show()
```

## **Current Rate of Motorization**


🏛️ The Impact and Relationship to the Diagnostic Study

Introducing a categorical tiering system to the Current Rate of Motorization (2016) moves your analysis beyond basic country rankings and provides direct, actionable answers to two of the most critical questions in Chapter 1.5 (The E-Mobility Journey in East Africa) and Chapter 7.0 (Challenges to EV Adoption):

1. It Establishes the Mandatory Regional Market Strategy

- The High-Density Tier (>20 vehicles / 1k people): This tier reveals where vehicle ownership is already culturally and economically entrenched. In these countries, the electric vehicle transition is fundamentally a fleet replacement game. Policy here must focus on aggressively flipping existing, highly polluting vehicle segments (like urban internal combustion Boda-Bodas) over to electric drivetrains.
- The Low-to-Medium Tiers (<20 vehicles / 1k people): In these environments, the vehicle market is still actively growing. The transition here is an infancy growth game. Instead of trying to replace old fleets, policies should aim to ensure that the millions of new vehicles entering the economy are electric from day one, preventing these countries from locking themselves into a reliance on dirty, secondhand fossil-fuel cars.

- Private charging network developers need high vehicle density to ensure their charging stations are consistently utilized and profitable.
```python
# Use your existing current_mot data frame variable directly
current_mot = current_mot.sort_values(
    by='Current rate of motorization (2016) (number of vehicles per 1000 people)', 
    ascending=False
)

# 1. Assign descriptive visual colors based on your strict numeric thresholds
def assign_color_tier(val):
    if val > 20:
        return '#0284C7'  # High (>20) - Ocean Blue
    elif 10 <= val <= 20:
        return '#EAB308'  # Medium (10-20) - Soft Gold
    else:
        return '#EF4444'  # Low (<10) - Coral Red

current_mot['Color'] = current_mot['Current rate of motorization (2016) (number of vehicles per 1000 people)'].apply(assign_color_tier)

# 2. Canvas and styling initialization
fig, ax = plt.subplots(figsize=(12, 7), dpi=100)
ax.set_facecolor('#F8FAFC')

# 3. Create clean horizontal bars matching your variable's exact columns
bars = ax.barh(
    current_mot['Country'], 
    current_mot['Current rate of motorization (2016) (number of vehicles per 1000 people)'], 
    color=current_mot['Color'], 
    edgecolor='#0F172A', 
    height=0.6, 
    zorder=3
)

# 4. Add shaded threshold category blocks behind the bars
ax.axvspan(0, 10, color='#FEE2E2', alpha=0.3, zorder=1)   # Low Zone Shading
ax.axvspan(10, 20, color='#FEF9C3', alpha=0.3, zorder=1)  # Medium Zone Shading
ax.axvspan(20, 55, color='#E0F2FE', alpha=0.3, zorder=1)  # High Zone Shading

# Add boundary lines and text annotations for the categories
ax.axvline(x=10, color='#94A3B8', linestyle=':', linewidth=1.5, zorder=2)
ax.axvline(x=20, color='#94A3B8', linestyle=':', linewidth=1.5, zorder=2)

# FIX: Isolate the upper bound [1] of the tuple before multiplying
y_max = ax.get_ylim()[1]

ax.text(5, y_max * 0.92, 'LOW DENSITY\n(<10 per 1000)', 
        ha='center', va='top', fontsize=9, color='#991B1B', weight='bold')
ax.text(15, y_max * 0.92, 'MEDIUM DENSITY\n(10-20 per 1000)', 
        ha='center', va='top', fontsize=9, color='#854D0E', weight='bold')
ax.text(37, y_max * 0.92, 'HIGH DENSITY\n(>20 per 1000)', 
        ha='center', va='top', fontsize=9, color='#0369A1', weight='bold')

# 5. Add exact data value annotations onto the tips of the bars
for bar in bars:
    width = bar.get_width()
    ax.text(
        width + 0.8, bar.get_y() + bar.get_height()/2, 
        f"{width:.1f}", 
        va='center', ha='left', color='#0F172A', weight='bold', size=11
    )

# 6. Professional formatting and label clean-up using your exact column text
ax.set_title('EAC Motorization Baseline & Market Segmentation\nClassifying Current Vehicle Fleet Density per 1,000 Inhabitants', 
             fontsize=14, pad=22, weight='bold', color='#1E293B')
ax.set_xlabel('Current rate of motorization  (number of vehicles per 1000 people)', fontsize=11, labelpad=12, weight='semibold', color='#334155')
ax.set_xlim(0, 55)

# Clean up layout borders
for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)
ax.grid(axis='x', linestyle='--', color='#E2E8F0', alpha=0.5, zorder=1)

plt.title("Current of Motorization", fontweight='bold')
plt.tight_layout()
plt.show()
```

## **Current Rate of Motorization Vs Projected Rate of Motorization**


**🏛️ The Impact and Relationship to the Diagnostic Study**

Analyzing Current vs. Projected Motorization Density with categorical tiering across the EAC countries introduces a critical, forward-looking dimension to your framework (Chapters 1.3, 1.5, and 16.0). It shifts the narrative from static baseline numbers to a dynamic growth model:

1. Quantifying the Infrastructure Catch-Up Velocity:
	- This multi-variable analysis exposes the structural timeline pressure facing energy and transport ministries. It demonstrates to policymakers that a country sitting in the "Low" or "Medium" category today cannot design its infrastructure based on current demand. They must model their grids to absorb the massive, incoming wave of vehicle volume as these markets rapidly scale into the "High" category by 2050.
2. Defending the Rationale for Upfront Fleet Electrification:
	- For non-technical audiences, this chart provides an immediate justification for sweeping regional e-mobility laws now. It proves visually that if countries wait until they cross into the "High Density" bracket to mandate electric vehicle adoption, they will have already allowed millions of dirty internal combustion engines to saturate their cities, locking themselves into decades of severe oil dependency and heavy urban pollution

```python
main = pd.read_sql('SELECT * FROM eac_df',conn)
```

```python
print(main[['Country',
            'Current Rate of Motorization (2016)',
            'Projected rate of motorization (2050)']].to_string(index=False))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --- 1. Coerce projected column to numeric (turns "Not Found" into NaN) ---
main['Projected rate of motorization (2050)'] = pd.to_numeric(
    main['Projected rate of motorization (2050)'], errors='coerce'
)
main['Current Rate of Motorization (2016)'] = pd.to_numeric(
    main['Current Rate of Motorization (2016)'], errors='coerce'
)

# --- 2. Category functions ---
def cat_current(v):
    if pd.isna(v): return 'Unknown'
    if v < 10:   return 'Low'
    if v <= 20:  return 'Medium'
    return 'High'

def cat_projected(v):
    if pd.isna(v): return 'Unknown'
    if v < 50:    return 'Low'
    if v <= 80:   return 'Medium'
    if v <= 100:  return 'High'
    return 'Very High'

df = main.copy()
df['Curr_Cat'] = df['Current Rate of Motorization (2016)'].apply(cat_current)
df['Proj_Cat'] = df['Projected rate of motorization (2050)'].apply(cat_projected)

# Sort by projected (NaN last)
df = df.sort_values('Projected rate of motorization (2050)', ascending=True, na_position='first')

# --- 3. Plot ---
y = np.arange(len(df))
h = 0.38

fig, ax = plt.subplots(figsize=(13, 7), dpi=110)
ax.set_facecolor('#F8FAFC')

bars_p = ax.barh(y + h/2, df['Projected rate of motorization (2050)'], h,
                 color='#0284C7', edgecolor='#0F172A', linewidth=1,
                 label='Projected (2050)')
bars_c = ax.barh(y - h/2, df['Current Rate of Motorization (2016)'], h,
                 color='#94A3B8', edgecolor='#0F172A', linewidth=1,
                 label='Current')

# Value + category labels
for bar, val, cat in zip(bars_p, df['Projected rate of motorization (2050)'], df['Proj_Cat']):
    label = f'{val:.0f}  ({cat})' if pd.notna(val) else 'No data'
    ax.text(bar.get_width() + 2,
            bar.get_y() + bar.get_height()/2,
            label, va='center', ha='left',
            color='#0369A1', weight='bold', size=9)

for bar, val, cat in zip(bars_c, df['Current Rate of Motorization (2016)'], df['Curr_Cat']):
    label = f'{val:.1f}  ({cat})' if pd.notna(val) else 'No data'
    ax.text(bar.get_width() + 2,
            bar.get_y() + bar.get_height()/2,
            label, va='center', ha='left',
            color='#475569', size=9)

# Threshold lines
for x in (10, 20):
    ax.axvline(x, color='#94A3B8', linestyle=':', linewidth=1.2)
for x in (50, 80, 100):
    ax.axvline(x, color='#0284C7', linestyle='--', linewidth=1.2, alpha=0.7)

ax.set_yticks(y)
ax.set_yticklabels(df['Country'], fontsize=11, weight='bold', color='#1E293B')
ax.set_xlabel('Motorization Rate (vehicles per 1,000 inhabitants)',
              fontsize=11, weight='semibold', color='#334155', labelpad=12)
ax.set_title('Current  vs Projected (2050) Motorization Rates by Country',
             fontsize=14, weight='bold', color='#1E293B', pad=20)

for s in ['top', 'right']:
    ax.spines[s].set_visible(False)
ax.grid(axis='x', linestyle='--', color='#E2E8F0', alpha=0.6)
ax.legend(loc='lower right', frameon=True, facecolor='white', edgecolor='#E2E8F0')

plt.tight_layout()
plt.show()
```


**🔍 Core Analytical Observations from the Output**

	- Kenya stands out as the ultimate regional market anchor. It is scaling rapidly from a baseline of 53.9 vehicles to a massive projection of 174 vehicles per 1,000 people, which represents a 3.2x market expansion.
	- Rwanda exhibits the highest growth velocity in the region, jumping from 15.2 (Medium density) straight to 126 per 1,000 individuals an astonishing 8.2x expansion.
	- Tanzania (109) and Uganda (106) both break past the 100-vehicle threshold, tightly mirroring each other's long-term scale.
2. The High-Density Convergence Zone:

## **Fuel Import Dependency**


**🏛️ The Impact and Relationship to the Diagnostic Study**

Analyzing Fossil Fuel Import Dependency captured via the column Fossil Fuel Dependence (Ratio of transportation fuel consumption to net imports by country) (2021) is the core argument for Chapter 2.0 (Rationale) and Chapter 2.3 (Challenges).

For Ministries of Finance, this variable shifts the project from an "environmental option" to a matter of macroeconomic survival:

1. Exposing the Foreign Exchange (Forex) Structural Leak:
	
	- Most EAC partner states do not produce crude oil and must spend massive reserves of scarce foreign currency to buy refined petroleum products. This variable tracks how much of that imported fuel is immediately burned up by the domestic transport sector. It demonstrates to central banks that transportation is a direct leak of national wealth. Swapping imported fuel for locally generated renewable electricity (hydro, geothermal, solar) allows that wealth to stay inside the country.

2. The Inflation and Logistic Shield (The 2026 Fuel Shock Context):
	
	- As of late 2026, fuel has spiked to an unprecedented 55% of total road transport costs in East Africa, up from 38% in 2021. This massive price surge directly drives up the cost of food and consumer goods across the region. Our framework uses this metric to prove that e-mobility is a vital economic shield decoupling transport costs from global oil volatility protects the regional economy from inflation.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Ensure your dataset is sorted directly using your active workspace variable
fuel_import = fuel_import.sort_values(by='Transport to Net Import Ratio (%)', ascending=False)

# 1. Canvas initialization and professional design
fig, ax = plt.subplots(figsize=(13, 8), dpi=100)
ax.set_facecolor('#F8FAFC')

# 2. Create high-contrast horizontal bars mapping your exact column name
bars = ax.barh(
    fuel_import['Country'], 
    fuel_import['Transport to Net Import Ratio (%)'], 
    color='#EF4444',  # Coral Red for heavy macroeconomic dependency risk
    edgecolor='#0F172A', 
    height=0.6, 
    zorder=3
)

# 3. Add a strict visual Macro Drain Guideline at the 75% threshold
ax.axvline(x=75.0, color='#DC2626', linestyle='--', linewidth=2, label='High Forex Leakage Threshold (75%)', zorder=2)

# Position category labels safely at the top of the chart canvas
y_limits = ax.get_ylim()
ax.text(76.5, y_limits[1] * 0.95, ' HIGH VULNERABILITY\nMACRO DRAIN ZONE (75%)', 
        ha='left', va='top', fontsize=9, color='#B91C1C', weight='bold')

# 4. Add exact numeric value labels safely onto the tips of the bars
for i, bar in enumerate(bars):
    width = bar.get_width()
    score = fuel_import['Dependence Score'].iloc[i]
    
    if not np.isnan(width):
        # Fallback handling to display score cleanly whether stored as float or string text
        score_label = f"{score:.1f}" if isinstance(score, (int, float, np.number)) else str(score)
        
        ax.text(
            width + 1.0, bar.get_y() + bar.get_height()/2, 
            f"{width:.1f}% (Risk: {score_label})", 
            va='center', ha='left', color='#0F172A', weight='bold', size=10
        )

# 5. Clean up layout typography and borders using your exact dataset columns
ax.set_title('EAC Macro-Financial Dependency Audit: Transport Petroleum Leakage\nProportion of Total Net National Petroleum Imports Consumed Exclusively by Transport Activities', 
             fontsize=14, pad=25, weight='bold', color='#1E293B')
ax.set_xlabel('Transport to Net Import Ratio (%)', fontsize=11, labelpad=12, weight='semibold', color='#334155')

# Dynamically set horizontal limit safely to leave plenty of room for value text
ax.set_xlim(0, max(fuel_import['Transport to Net Import Ratio (%)'].max(), 75.0) * 1.35)

for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)
ax.grid(axis='x', linestyle='--', color='#E2E8F0', alpha=0.5, zorder=1)
ax.legend(loc='lower right', frameon=True, facecolor='white', edgecolor='#E2E8F0')

plt.tight_layout()
plt.show()
```

## **Policy Inventory**


**🏛️ The Impact and Relationship to the Diagnostic Study**

This table changes the narrative of your legal audit by shifting from simple lists to a clear measure of regulatory structure and implementation velocity:

1. Visualizing the Shift from Fiscal Carrots to Regulatory Sticks (Chapter 5.0)

- The Impact: Early-stage markets typically start with fiscal incentives (like import duty cuts) to make vehicles affordable. As the market matures, they transition to regulatory frameworks (like charging grid rules or combustion engine bans). Aggregating the instrument_type column across your country timelines visually maps this evolution, showing non-technical stakeholders exactly which states are leading this regulatory transition.

2. Identifying the Regional Policy Fragmentation Gap (Chapter 3.0)

- The Impact: For an economic bloc to thrive, vehicle standards and charging rules must match across borders. Counting active policies by category across the eight nations exposes fragmentation. It highlights where cross-border trade bottlenecks will occur such as an electric truck facing different charging connector guidelines or battery safety rules when crossing from Kenya into Uganda.


```python
conn = sqlite3.connect('emobility_cleaned.db')  # reopen
policy = pd.read_sql("SELECT * FROM policy_inventory", conn)
```

```python

policy
```

```python
policy.columns
```

```python
import pandas as pd
import numpy as np

# Ensure your active workspace variable is sorted chronologically by year
policy_sorted = policy.sort_values(by=['year', 'country_iso']).copy()

# Print clean structural title block
print("=" * 140)
print(f"{'EAC REGIONAL E-MOBILITY POLICY REGISTRY TIMELINE & DATA SOURCE AUDIT':^140}")
print("=" * 140)
print(f"{'YEAR':<5} | {'ISO':<5} | {'SEGMENT':<9} | {'POLICY TYPE & INSTRUMENT':<40} | {'STRATEGIC INFRASTRUCTURE IMPACT':<50} | {'PRIMARY DATA SOURCE'}")
print("-" * 140)

# Loop through the text rows to print a clean visual timeline table
for i, row in policy_sorted.iterrows():
    year = str(row['year'])
    country = str(row['country_iso'])
    segment = str(row['vehicle_segment']).upper()
    instrument = str(row['enacted_policy_instrument'])
    desc = str(row['description'])
    source = str(row['data_source'])
    ins_type = str(row['instrument_type']).lower()
    
    # Skip placeholder rows with zero policy action to keep the focus on active steps
    if "no ev policy" in instrument.lower():
        continue
        
    # Apply striking visual anchors based on the legal mechanism type
    if ins_type == 'fiscal':
        marker = "🟢 [TAX] "
    elif 'ban' in instrument.lower() or 'mandate' in instrument.lower():
        marker = "🔴 [BAN] "
    else:
        marker = "🔵 [REG] "
        
    # Clean up long text strings so they fit beautifully on standard displays
    full_instrument = marker + instrument
    clean_instrument = full_instrument[:38] + "..." if len(full_instrument) > 38 else full_instrument
    clean_desc = desc[:48] + "..." if len(desc) > 48 else desc
    clean_source = source[:22] + "..." if len(source) > 22 else source
    clean_segment = segment[:9]
    
    print(f"{year:<5} | {country:^5} | {clean_segment:<9} | {clean_instrument:<40} | {clean_desc:<50} | {clean_source}")

print("=" * 140)
print("🟢 Fiscal = Tax & Duty Relief  |  🔵 Regulatory = System Grid Standards  |  🔴 Mandates = Binding Phase-outs & Bans")
print("=" * 140)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# 1. Generate a clean cross-tabulation count from your active variable 'policy'
policy_counts = pd.crosstab(
    index=policy['country_iso'], 
    columns=policy['instrument_type']
).reset_index()

# Handle missing column keys gracefully if a type isn't present in the slice
for col in ['fiscal', 'regulatory']:
    if col not in policy_counts.columns:
        policy_counts[col] = 0

# Sort by total volume of policy structures to align with your core economic zones
policy_counts['Total_Instruments'] = policy_counts['fiscal'] + policy_counts['regulatory']
policy_counts = policy_counts.sort_values(by='Total_Instruments', ascending=False)

# 2. Canvas initialization and professional design
fig, ax = plt.subplots(figsize=(12, 6), dpi=100)
ax.set_facecolor('#F8FAFC')

y_indices = np.arange(len(policy_counts))
bar_height = 0.35

# 3. Plot side-by-side grouped horizontal bars 
bars_fiscal = ax.barh(y_indices + bar_height/2, policy_counts['fiscal'], bar_height, 
                       label='Fiscal Incentives (Tax/Duty Relief)', color='#10B981', edgecolor='#0F172A', linewidth=1)

bars_reg = ax.barh(y_indices - bar_height/2, policy_counts['regulatory'], bar_height, 
                    label='Regulatory Frameworks (Mandates/Guidelines)', color='#0284C7', edgecolor='#0F172A', linewidth=1)

# 4. Add smart integer counts onto the tips of the bars for non-tech clarity
for bar in bars_fiscal:
    width = bar.get_width()
    if width > 0:
        ax.text(width + 0.05, bar.get_y() + bar.get_height()/2, f"{int(width)}", 
                va='center', ha='left', color='#065F46', weight='bold', size=10)

for bar in bars_reg:
    width = bar.get_width()
    if width > 0:
        ax.text(width + 0.05, bar.get_y() + bar.get_height()/2, f"{int(width)}", 
                va='center', ha='left', color='#0369A1', weight='bold', size=10)

# 5. Clean up layout, labels, and borders
ax.set_title('EAC Regulatory Landscape Diagnostic: Policy Instrument Mix\nQuantifying Enacted Fiscal Incentives versus Binding Technical and Environmental Regulations', 
             fontsize=13, pad=22, weight='bold', color='#1E293B')
ax.set_xlabel('Count of Active Policy Instruments Sourced in Registry', fontsize=11, labelpad=12, weight='semibold', color='#334155')
ax.set_yticks(y_indices)

# FIX: Map the ISO codes to actual country names cleanly for a non-tech audience
iso_to_name = {
    'KE': 'Kenya', 'TZ': 'Tanzania', 'UG': 'Uganda', 'RW': 'Rwanda', 
    'BI': 'Burundi', 'SS': 'South Sudan', 'SO': 'Somalia', 'CD': 'DR Congo'
}
full_country_names = [iso_to_name.get(iso, iso) for iso in policy_counts['country_iso']]
ax.set_yticklabels(full_country_names, fontsize=11, weight='bold', color='#1E293B')

# Force clean integer bounds for a non-technical audience
max_val = max(policy_counts['fiscal'].max(), policy_counts['regulatory'].max())
ax.set_xlim(0, max_val + 1)
ax.xaxis.get_major_locator().set_params(integer=True)

for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)
ax.grid(axis='x', linestyle='--', color='#E2E8F0', alpha=0.5, zorder=1)
ax.legend(loc='lower right', frameon=True, facecolor='white', edgecolor='#E2E8F0')

plt.tight_layout()
plt.show()

```

Analyzing Fiscal Tariffs specifically comparing the import duties, excise taxes, value-added taxes (VAT), and registration fees applied to Internal Combustion Engine (ICE) vehicles versus Electric Vehicles (EVs) reaches the absolute core of your diagnostic study.
This dataset underpins Chapter 4.0 (Policy Development Process), Chapter 8.0 (Incentives), and Chapter 9.0 (Importation of EVs). Before you run any code or look at the rows, you must frame this data around its critical structural impacts on the region:

As we established in our Group A income metrics, individual consumer purchasing power across the EAC hovers within a tight baseline tier ($1,000–$2,000 per capita in the strongest core markets). Because electric vehicles carry an intense manufacturing cost premium due to their advanced lithium battery chemistries, they are naturally more expensive upfront than a legacy gasoline equivalent.

The Impact: Tariffs are the primary tool a government has to artificially force price parity. This dataset calculates the Tariff Advantage Delta across borders. The Impact: This creates a severe inversion that penalizes local value chains. It makes it cheaper to import finished foreign vehicles than to build local assembly, battery-swapping, and component-manufacturing plants within the EAC, contradicting the industrial growth goals of the EAC Treaty.

The eight EAC partner states share a theoretical Common External Tariff (CET), but national deviations, local tax exemptions, and different VAT applications create steep tariff borders between contiguous neighbors.

The Impact: If Kenya offers a 10% total tax burden on electric motorcycles while neighboring Uganda or Tanzania applies a 45% standard tariff, it creates a massive price distortion. Waiving this final multiplier layer will instantly lower the asset purchase price for low-income operators, accelerating transport transition without causing major losses to national tax revenues.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Use your existing variable 'fiscal' directly
# 1. Clean and convert columns to floats to protect against mixed text/float type mismatches
for col in ['Import Duty (ICE)', 'Import Duty (EV)', 'Excise Duty (ICE)', 'Excise Duty (EV)', 'VAT Rate (ICE)', 'VAT Rate (EV)']:
    # FIX: Safely parse the column directly from the 'fiscal' dataframe
    fiscal[col] = pd.to_numeric(fiscal[col], errors='coerce').fillna(0)
    
    # If the column values are stored as whole numbers (e.g. 16 for 16%), convert to decimals
    if fiscal[col].max() > 1.0:
        fiscal[col] = fiscal[col] / 100.0

# 2. Compute the True Cascading Cumulative Port Tax Multiplier mathematically
# Formula accounts for standard compounding customs architecture: (1 + Import) * (1 + Excise) * (1 + VAT) - 1
fiscal['Total_ICE_Tax_Pct'] = ((1 + fiscal['Import Duty (ICE)']) * (1 + fiscal['Excise Duty (ICE)']) * (1 + fiscal['VAT Rate (ICE)']) - 1) * 100
fiscal['Total_EV_Tax_Pct'] = ((1 + fiscal['Import Duty (EV)']) * (1 + fiscal['Excise Duty (EV)']) * (1 + fiscal['VAT Rate (EV)']) - 1) * 100

# Calculate the precise Tariff Advantage Delta (Positive means EV is taxed less)
fiscal['Tariff_Advantage_Delta'] = fiscal['Total_ICE_Tax_Pct'] - fiscal['Total_EV_Tax_Pct']
fiscal = fiscal.sort_values(by='Tariff_Advantage_Delta', ascending=False)

# 3. Canvas initialization and professional design
fig, ax = plt.subplots(figsize=(12, 7), dpi=100)
ax.set_facecolor('#F8FAFC')

y_indices = np.arange(len(fiscal))
bar_height = 0.35

# 4. Plot True Cumulative Compounded Border Taxes side-by-side
bars_ice = ax.barh(y_indices - bar_height/2, fiscal['Total_ICE_Tax_Pct'], bar_height, 
                    label='Total Compounded Tax Burden: ICE', color='#64748B', edgecolor='#0F172A', linewidth=1)

bars_ev = ax.barh(y_indices + bar_height/2, fiscal['Total_EV_Tax_Pct'], bar_height, 
                   label='Total Compounded Tax Burden: EV', color='#10B981', edgecolor='#0F172A', linewidth=1)

# 5. Add exact compounded values onto the tips of the bars for non-tech clarity
for bar in bars_ice:
    width = bar.get_width()
    if width > 0:
        ax.text(width + 1.0, bar.get_y() + bar.get_height()/2, f"{width:.1f}%", 
                va='center', ha='left', color='#475569', size=9)

for bar in bars_ev:
    width = bar.get_width()
    if width > 0:
        ax.text(width + 1.0, bar.get_y() + bar.get_height()/2, f"{width:.1f}%", 
                va='center', ha='left', color='#065F46', weight='bold', size=10)

# Professional typography and label clean-up using your full country name map
ax.set_title('EAC Port Custom Diagnostics: Total Cumulative Vehicle Tax Burdens\nQuantifying Compounded Port Duties, Excise Taxes, and VAT Applications (ICE vs. Electric Vehicles)', 
             fontsize=13, pad=25, weight='bold', color='#1E293B')
ax.set_xlabel('True Cumulative Compounded Port Tax Burden (%)', fontsize=11, labelpad=12, weight='semibold', color='#334155')
ax.set_yticks(y_indices)

# Map the exact country text cleanly
iso_to_name = {
    'KE': 'Kenya', 'TZ': 'Tanzania', 'UG': 'Uganda', 'RW': 'Rwanda', 
    'BI': 'Burundi', 'SS': 'South Sudan', 'SO': 'Somalia', 'CD': 'DR Congo',
    'Kenya': 'Kenya', 'Tanzania': 'Tanzania', 'Uganda': 'Uganda', 'Rwanda': 'Rwanda',
    'Burundi': 'Burundi', 'South Sudan': 'South Sudan', 'Somalia': 'Somalia', 'DR Congo': 'DR Congo'
}
full_country_names = [iso_to_name.get(c, c) for c in fiscal['Country']]
ax.set_yticklabels(full_country_names, fontsize=11, weight='bold', color='#1E293B')

# Expand horizontal limit safely to leave plenty of room for value tags
ax.set_xlim(0, max(fiscal['Total_ICE_Tax_Pct'].max(), fiscal['Total_EV_Tax_Pct'].max()) + 15)

for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)
ax.grid(axis='x', linestyle='--', color='#E2E8F0', alpha=0.5, zorder=1)
ax.legend(loc='upper right', frameon=True, facecolor='white', edgecolor='#E2E8F0')

plt.tight_layout()
plt.show()

```


**🔍 Core Analytical Observations from the Output**

1. Kenya's Aggressive "Incentive Delta" Leadership:
	
 Kenya has created the most significant tax advantage for clean energy vehicles in the entire region. While standard internal combustion engines (ICE) face a massive 99.7% compounded tax burden at the port, electric vehicles (EVs) are taxed at just 27.6%. This marks a massive 72.1% tax discount, making Kenya the absolute frontrunner in using fiscal policy to force upfront price parity.

The Impact: Public charging networks require massive upfront capital expenditures (CapEx) from private operators. ## **HS Codes**

What it means: HS Codes are internationally standardized numerical codes used by customs authorities (like the Kenya Revenue Authority or Tanzania Revenue Authority) to classify traded goods and apply correct import duties, excise taxes, and VAT.

The Diagnostic Reality: Traditional customs frameworks were written for Internal Combustion Engine (ICE) vehicles. A clean e-mobility matrix must assign discrete HS codes distinguishing:
	
  - CBU (Completely Built Units): Fully assembled electric cars. High-tariff items meant to protect local assembly lines.
  - CKD (Completely Knocked Down) Kits: Loose parts imported to be welded and bolted together inside domestic factories. Usually zero-rated or low-tariff to spur local jobs.
  - Lithium Cells & Drivetrains: Separated raw technology components. Crucial for local battery-pack manufacturers.
  - EVSE (Electric Vehicle Supply Equipment): Public fast chargers and domestic wallboxes.

2. Adopted Safety and Performance Standards: The Technical Shield

What it means: These are the mandatory engineering certifications enforced by national standards bureaus (like KEBS in Kenya, TBS in Tanzania, or RSB in Rwanda) ensuring that imported technology does not pose fire, environmental, or electrocution hazards.

The Diagnostic Reality: Instead of reinventing the wheel, the matrix tracks how countries import international benchmarks:
  - UN ECE R100: The strict United Nations standard governing the safety requirements of lithium-ion traction batteries (vibration, thermal shock, mechanical impact, and short-circuit protection).
  - Battery End-of-Life (EOL): E-waste and toxic disposal regulations dictating how batteries are safely recycled or repurposed inside the EAC when their health degrades.

3. Charging Protocols: The Interoperability Language

What it means: This dictates the physical shape of the charging plug and the digital communication protocol passing between the vehicle and the charging station.

The Diagnostic Reality: Different global manufacturing regions utilize different plug layouts:
 - CCS2 (Combined Charging System Type 2): The standard European plug layout, widely adopted in Kenya, Uganda, and Rwanda for commercial buses and private cars.
 - GB/T: The national Chinese charging standard. Massive entry footprint across East Africa due to the high volume of affordable Chinese e-motorcycles and three-wheelers.
 - CHAdeMO: The legacy Japanese fast-charging standard, occasionally seen on second-hand imported Japanese EVs (like older Nissan Leafs).

**🏛️ The Impact and Relationship to the Study**

This matrix affects your diagnostic study in three distinct operational areas:

1. It Exposes the Risk of "Technology Fragmentation" (Chapter 3.0 & 5.0)

- The Relationship: If individual EAC states adopt conflicting plug standards for instance, if Kenya mandates CCS2 for heavy trucks while Tanzania enforces GB/T vehicles cannot easily cross regional border entry points.
- The Impact: The matrix acts as a diagnostic tool showing where standard misalignment breaks cross-border logistics. Your study uses this to argue for a harmonized regional standard agreement, allowing an electric freight truck departing from the Port of Mombasa to seamlessly recharge using the same physical connection points all the way to Kampala or Kigali.

2. It Exposes the "Fiscal Leakage" vs. Local Assembly Paradox (Chapter 5.2)

- The Relationship: Revenue authorities cannot accurately issue tax breaks if their domestic tariff books lack specific codes for EV components. Without clear HS classifications, a customs officer might accidentally tax an electric motor or a lithium-ion battery cell at the same high luxury rate as an ICE replacement engine.
- The Impact: This visually justifies your study’s policy recommendations. By analyzing the current classification gaps, the framework recommends that newer or slower-tier EAC entrants immediately update their customs tariff structures to distinctly isolate EV components from ICE parts, giving domestic assemblers a predictable, low-tariff manufacturing environment.

3. It Prevents sub-standard Technology Dumping (Chapter 8.0 & 9.0)

- The Relationship: Fragile or newer EAC markets lacking active safety enforcement protocols run the risk of becoming dumping grounds for cheap, low-grade, or second-hand batteries that do not meet standard thermal management benchmarks.
- The Impact: This shapes the technical assistance sequencing in your roadmap. The diagnostic matrix shows that before a country launches large-scale consumer vehicle consumer subsidies, it must first establish basic laboratory testing and homologation mechanisms at its standards bureaus (like KEBS or RSB) to verify international UN ECE certifications on incoming tech shipments.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Set clean presentation styling 
sns.set_theme(style="whitegrid")
plt.rcParams["font.family"] = "sans-serif"

# =====================================================================
# LAYOUT 1: Fixed, Visible Matplotlib Data Table (With Explicit Bounding Box)
# =====================================================================
def generate_visible_table(df):
    table_cols = [
        "country_name",
        "standards_bureau_code",
        "battery_safety_benchmark",
        "primary_ac_plug",
        "primary_dc_fast_plug"
    ]
    table_data = df[table_cols].copy()
    
    # Increase figure size and create an adjusted canvas space
    fig, ax = plt.subplots(figsize=(14, 5))
    ax.axis("off")  # Completely turn off normal plot axes
    
    # Render table with explicit bounding box coordinates [left, bottom, width, height]
    # This prevents the table from turning blank or shifting outside the frame!
    table_obj = ax.table(
        cellText=table_data.values,
        colLabels=[col.replace("_", " ").title() for col in table_data.columns],
        cellLoc="center",
        loc="center",
        bbox=[0.0, 0.1, 1.0, 0.8]  
    )
    
    table_obj.auto_set_font_size(False)
    
    # Format properties cell by cell safely
    for (row, col), cell in table_obj.get_celld().items():
        cell.set_fontsize(10)
        if row == 0:
            cell.set_text_props(weight="bold", color="white")
            cell.set_facecolor("#2c3e50")  # Dark corporate navy header
        else:
            if row % 2 == 0:
                cell.set_facecolor("#f8f9fa")  # Soft alternating row tint
            if col == 0:
                cell.set_text_props(weight="bold")  # Bold country names column
                
    plt.title("EAC E-Mobility Interface and Technical Registry Table", fontsize=14, weight="bold", pad=20)
    plt.show()

# =====================================================================
# LAYOUT 2: Standardized Plug Protocol Distribution (Bar Chart)
# =====================================================================
def generate_plug_distribution_bar(df):
    # Melt the plug protocol columns to create a tidy format for a side-by-side comparison
    plug_df = df.melt(
        id_vars=["country_name"],
        value_vars=["primary_ac_plug", "primary_dc_fast_plug"],
        var_name="charging_interface_type",
        value_name="plug_protocol_standard"
    )
    
    # Clean out any empty strings or unregistered plug entries to keep focus sharp
    plug_df = plug_df[plug_df["plug_protocol_standard"].str.lower().str.strip() != "unregulated"]
    
    plt.figure(figsize=(13, 6))
    
    # Generate grouped distribution comparison bar plot
    ax = sns.countplot(
        x="plug_protocol_standard",
        hue="charging_interface_type",
        data=plug_df,
        palette=["#3498db", "#e74c3c"]  # Light blue for AC, Red for DC Fast
    )
    
    # Force integer ticks on the Y axis since these are raw counts of country profiles
    plt.gca().yaxis.get_major_locator().set_params(integer=True)
    
    plt.title(
        "Distribution of Adopted EV Charging Port Standards Across the EAC\n"
        "(Counting Standardized Deployments Across Partner States)",
        fontsize=14, weight="bold", pad=15
    )
    plt.xlabel("Plug Protocol Standard", fontsize=11, labelpad=10)
    plt.ylabel("Number of EAC Countries Using Standard", fontsize=11)
    
    # Clean up legend labels
    plt.legend(
        labels=["Primary AC Port", "Primary DC Fast Port"],
        title="Interface Classification", loc="upper right"
    )
    plt.xticks(rotation=80)
    plt.tight_layout()
    plt.show()

# =====================================================================
# RUN THE VISUALIZATIONS
# =====================================================================
generate_visible_table(hs_codes)
generate_plug_distribution_bar(hs_codes)
```

While a 20 MW surplus in Somalia or Burundi represents a 16–20% buffer on paper, in reality, it is smaller than the output of a single mid-sized generation turbine. Your framework must use this to show that frontier nations cannot rely on standard percentage metrics; they require a minimum absolute megawatt threshold to safely handle unplanned outages without triggering widespread blackouts.

2. Exposing Opportunities for Regional Power-Pool Arbitrage (Chapter 4.1):

The immense disparity between localized demand and generation assets reveals a massive opportunity for cross-border power trading. For instance, the 1,100 MW surplus in the DR Congo and the 900 MW surplus in Kenya sit immediately adjacent to frontier nations like Rwanda and South Sudan that are choking on sub-50 MW margins.
```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Sort from cleanest to thermal-dominant for a professional left-to-right presentation flow
power['Total_Clean_Ratio'] = power['generation_hydro_pct'] + power['generation_solar_pct'] + power['generation_wind_pct'] + power['generation_other_pct']
power = power.sort_values(by='Total_Clean_Ratio', ascending=False)

fig, ax = plt.subplots(figsize=(13, 7), dpi=100)
ax.set_facecolor('#F8FAFC')

x_indices = np.arange(len(power))
bar_width = 0.55

# Scaling check: convert math decimals to whole percentages (0 to 100) if required
hydro = power['generation_hydro_pct'] * 100 if power['generation_hydro_pct'].max() <= 1.0 else power['generation_hydro_pct']
solar = power['generation_solar_pct'] * 100 if power['generation_solar_pct'].max() <= 1.0 else power['generation_solar_pct']
wind = power['generation_wind_pct'] * 100 if power['generation_wind_pct'].max() <= 1.0 else power['generation_wind_pct']
other = power['generation_other_pct'] * 100 if power['generation_other_pct'].max() <= 1.0 else power['generation_other_pct']
thermal = power['generation_thermal_pct'] * 100 if power['generation_thermal_pct'].max() <= 1.0 else power['generation_thermal_pct']

# Generate vertical stacked columns sequentially going upward
ax.bar(x_indices, hydro, label='Hydro-Electric', color='#059669', edgecolor='#0F172A', width=bar_width)
ax.bar(x_indices, solar, bottom=hydro, label='Solar PV', color='#F59E0B', edgecolor='#0F172A', width=bar_width)
ax.bar(x_indices, wind, bottom=hydro+solar, label='Wind Power', color='#06B6D4', edgecolor='#0F172A', width=bar_width)
ax.bar(x_indices, other, bottom=hydro+solar+wind, label='Other Clean (Geothermal/Biomass)', color='#10B981', edgecolor='#0F172A', width=bar_width)
ax.bar(x_indices, thermal, bottom=hydro+solar+wind+other, label='Thermal (Fossil Fuel / Oil)', color='#EF4444', edgecolor='#0F172A', width=bar_width)

ax.set_title('EAC Generation Asset Mix: Sourcing the Fleet Electricity Load\nProportion of National Electricity Generation Sourced from Renewable vs. Fossil Fuel Assets', 
             fontsize=13, pad=22, weight='bold', color='#1E293B')
ax.set_ylabel('Proportion of National Electricity Generation Mix (%)', fontsize=11, labelpad=12, weight='semibold', color='#334155')
ax.set_xticks(x_indices)

iso_to_name = {
    'KE': 'Kenya', 'TZ': 'Tanzania', 'UG': 'Uganda', 'RW': 'Rwanda', 
    'BI': 'Burundi', 'SS': 'South Sudan', 'SO': 'Somalia', 'CD': 'DR Congo',
    'Kenya': 'Kenya', 'Tanzania': 'Tanzania', 'Uganda': 'Uganda', 'Rwanda': 'Rwanda',
    'Burundi': 'Burundi', 'South Sudan': 'South Sudan', 'Somalia': 'Somalia', 'DR Congo': 'DR Congo'
}
ax.set_xticklabels([iso_to_name.get(c, c) for c in power['country_iso']], fontsize=11, weight='bold', color='#1E293B', rotation=15)
ax.set_ylim(0, 100)

ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=5, frameon=True, facecolor='white', edgecolor='#E2E8F0')
for spine in ['top', 'right']: ax.spines[spine].set_visible(False)
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```

## **Grid Reliability**


🏛️ The Infrastructure & Operational Impact of Grid Reliability

Analyzing Grid Reliability—focusing on metrics like the System Average Interruption Duration Index (SAIDI) or the average annual hours a nation spends without electricity—is the ultimate operational reality check for Chapter 11.0 (Energy and Charging Infrastructure).
Before diving into your specific data columns, we must anchor this indicator around its direct structural impacts on the transport transition:

1. It Identifies the "Range Anxiety" and Charger Downtime Bottleneck

If a country possesses massive electricity generation surpluses but suffers from a fragile local distribution grid, charging stations will experience frequent, unpredictable blackouts.
• The Impact: Frequent outages turn public fast-charging networks into a major liability for commercial operators. If a Boda-Boda rider pulls up to a battery-swap station only to find it offline due to a localized blackout, their daily income is instantly cut off. This data mathematically tracks Infrastructure Operational Risk, proving whether developers must invest in expensive battery backstops and solar arrays just to keep their charging hardware running.

2. It Dictates the Charging Strategy: Decentralized Overnight vs. High-Speed Peak

• The Impact: In markets with low grid reliability (high blackout hours), the framework cannot recommend high-speed highway fast-chargers that pull massive, instant spikes of power. Instead, the data forces policy to mandate Decentralized Overnight Depot Charging. This allows vehicles to charge slowly during off-peak hours when the distribution network is under the least amount of stress, using smart energy management to bypass local reliability weaknesses.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Safely parse text columns to numeric, replacing text placeholders with NaN
grid['Percent of firms experiencing electrical outages (%)'] = pd.to_numeric(
    grid['Percent of firms experiencing electrical outages (%)'], errors='coerce'
)

# Sort from lowest to highest for a clean horizontal ascending layout flow
grid_plot1 = grid.sort_values(by='Percent of firms experiencing electrical outages (%)', ascending=True)

fig, ax = plt.subplots(figsize=(13, 7), dpi=100)
ax.set_facecolor('#F8FAFC')

y_indices = np.arange(len(grid_plot1))

# Create clean horizontal bars for expanded label legibility
bars = ax.barh(
    y_indices, 
    grid_plot1['Percent of firms experiencing electrical outages (%)'], 
    color='#EF4444', edgecolor='#0F172A', height=0.6, zorder=3
)

# Map ISO codes to full country names cleanly
iso_to_name = {
    'KE': 'Kenya', 'TZ': 'Tanzania', 'UG': 'Uganda', 'RW': 'Rwanda', 
    'BI': 'Burundi', 'SS': 'South Sudan', 'SO': 'Somalia', 'CD': 'DR Congo',
    'Kenya': 'Kenya', 'Tanzania': 'Tanzania', 'Uganda': 'Uganda', 'Rwanda': 'Rwanda',
    'Burundi': 'Burundi', 'South Sudan': 'South Sudan', 'Somalia': 'Somalia', 'DR Congo': 'DR Congo'
}

# Add text values safely on the tips of the bars, marking text flags where data is missing
for i, bar in enumerate(bars):
    width = bar.get_width()
    country_raw = grid_plot1['Country'].iloc[i]
    
    if np.isnan(width):
        ax.text(1.5, bar.get_y() + bar.get_height()/2, " No Available Data", 
                va='center', ha='left', color='#B91C1C', weight='bold', size=10)
    else:
        ax.text(width + 1.0, bar.get_y() + bar.get_height()/2, f"{width:.1f}%", 
                va='center', ha='left', color='#0F172A', weight='bold', size=10)

ax.set_title('Widespread Grid Instability: Proportion of Commercial Operations Affected by Outages', 
             fontsize=13, pad=22, weight='bold', color='#1E293B')
ax.set_xlabel('Percent of firms experiencing electrical outages (%)', fontsize=11, labelpad=12, weight='semibold', color='#334155')
ax.set_yticks(y_indices)
ax.set_yticklabels([iso_to_name.get(c, c) for c in grid_plot1['Country']], fontsize=11, weight='bold', color='#1E293B')

# Set horizontal bounds safely with padding
max_val = grid_plot1['Percent of firms experiencing electrical outages (%)'].max()
ax.set_xlim(0, (max_val if not np.isnan(max_val) else 100) * 1.15)

for spine in ['top', 'right']: ax.spines[spine].set_visible(False)
ax.grid(axis='x', linestyle='--', alpha=0.5, zorder=1)
plt.tight_layout()
plt.show()

```

```python
import matplotlib.pyplot as plt

# Creates a clean 2x4 matrix layout of individual donut charts
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.flatten()

colors_donut = ["#218c74", "#b33939"] # Clean vs Dirty

for i, country in enumerate(egc["Country"]):
    clean_val = egc.loc[i, "% Clean"]
    dirty_val = egc.loc[i, "% Dirty"]
    shares = [clean_val, dirty_val]
    
    wedges, texts, autotexts = axes[i].pie(
        shares, labels=["", ""], autopct='%1.1f%%', 
        startangle=90, colors=colors_donut, pctdistance=0.75,
        textprops=dict(color="white", weight="bold", size=10)
    )
    
    # Drop labels for 0% portions to maintain clean legibility
    for autotext, share in zip(autotexts, shares):
        if share == 0:
            autotext.set_text("")
            
    # Punch out center to create premium donut framework
    centre_circle = plt.Circle((0,0), 0.55, fc='white')
    axes[i].add_artist(centre_circle)
    axes[i].set_title(f"{country}", weight="bold", size=12)

fig.suptitle("Proportional EAC Grid Environmental Composition (% Clean vs % Dirty)", fontsize=16, weight="bold")
fig.legend(labels=["% Clean (Renewables + Nuclear)", "% Dirty (Fossil Fuels)"], loc="lower center", ncol=2, bbox_to_anchor=(0.5, 0.02))
plt.tight_layout(rect=[0, 0.05, 1, 0.95])
plt.show()
```

2. The Mandatory Off-Grid Solar Decoupling Pivot (Chapter 4.1):

Conversely, for carbon-locked networks like Somalia (92.9% dirty), South Sudan (69.4% dirty), and Tanzania (67.1% dirty), standard grid-tied EV charging is counter-productive to national decarbonization goals. Our diagnosis framework must use these high "dirty" percentages to mandate a Decoupled Infrastructure Directive.
```python
import matplotlib.pyplot as plt

# Sort on clean energy percentage for proper scale progression
egc_sorted_score = egc.sort_values(by="% Clean", ascending=True)

plt.figure(figsize=(11, 5.5))

# Define clean color palette strings
color_low = "#ee5253"     # Red
color_medium = "#ff9f43"  # Orange
color_high = "#10ac84"    # Green

# Dynamically build color and label vectors directly from the percentages
hue_colors = []
tier_labels = []

for pct in egc_sorted_score["% Clean"].values:
    if pct < 33.0:
        hue_colors.append(color_low)
        tier_labels.append("Low")
    elif pct <= 66.0:
        hue_colors.append(color_medium)
        tier_labels.append("Medium")
    else:
        hue_colors.append(color_high)
        tier_labels.append("High")

# Build the horizontal scorecard graph
bars = plt.barh(egc_sorted_score["Country"], egc_sorted_score["% Clean"], color=hue_colors, edgecolor="black", height=0.6)

# Overlay policy thresholds on the graph layout
plt.axvline(x=33.0, color=color_low, linestyle=":", alpha=0.8, linewidth=2, label="Low Threshold (<33%)")
plt.axvline(x=66.0, color=color_medium, linestyle=":", alpha=0.8, linewidth=2, label="High Threshold (>66%)")

plt.title("EAC E-Mobility Strategic Decarbonization Readiness Scoring",fontweight='bold')
plt.xlabel("Clean Power Generation Proportional Base (%)")
plt.ylabel("Country")

# Right-aligned text anchors using the freshly generated lists
for bar, label, color in zip(bars, tier_labels, hue_colors):
    width = bar.get_width()
    plt.text(width + 1.5, bar.get_y() + bar.get_height()/2, f"Tier: {label}", va="center", weight="bold", color=color)

plt.xlim(0, 115)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```


**🔍 Core Analytical Observations from the Output**

Five out of the eight nations cross the critical >66% High Threshold, establishing an immediate green-light zone for electric vehicle grid integration. DR Congo and Uganda lead the cohort near the top of the scale (exceeding 94% clean base), while Burundi, Rwanda, and Kenya form a highly secure sub-cluster ranging between 80% and 86% clean power generation. In these five economies, the structural foundation is already in place to ensure that fleet electrification translates directly into deep, immediate lifecycle carbon reductions.

2. The High Threshold Gap (The Empty Medium-Tier Zone):

3. The Low-Tier Frontier Carbon Traps (Tanzania, South Sudan, & Somalia):

## **Daily Operations**

Analyzing Daily Operations represents the absolute baseline of micro-level diagnostics. While trip-level data captures individual route efficiencies, the daily operational scale reflects the actual economic reality of a Boda-Boda rider’s life. A rider does not balance their budget trip-by-trip; they balance it at the end of the day when they return home to their family.

By evaluating data aggregated at the daily operational level, we can isolate the daily fixed costs from the highly fluid variable costs, unlocking distinct diagnostic insights for both ICE and EV platforms:

- For ICE Motorbikes (The Volatile OpEx Trap): Daily operations expose the harsh reality of the "fuel hand-to-mouth" cycle. A rider's daily profitability is entirely hostage to the fluctuating retail price of petrol at local fuel stations, micro-maintenance leakages (daily oil top-ups, punctures), and unpredictable local route pricing.

- For EV Motorbikes (The Predictable Subscription Model): Daily operations capture the performance of the battery-swapping or smart-charging infrastructure. Instead of volatile fuel purchases, the EV rider encounters a more stabilized daily structure often framed around flat-rate battery lease fees, pay-as-you-swap packages, or managed overnight charging tariffs.

Ultimately, mapping out daily operations allows us to establish the Daily Net Margin. This metric is the single most vital indicator for determining whether an e-mobility shift successfully protects, or actively improves, a rider’s daily livelihood across the EAC region. ## **Trip**

While daily operations provide the macro-level ledger of a rider’s financial survival, Trip Operations shift the diagnostic lens into high-definition telemetry. A daily summary aggregates the results, but trip-level data exposes the exact behavioral choices, routing efficiencies, and spatial restrictions that create those results.

By analyzing data segmented at the individual trip level, we move away from simple accounting and begin mapping the physical physics of the Boda-Boda economy. This micro-telemetry isolates operational friction points that are invisible at the daily level, offering critical insights for both platforms:

1. For ICE Motorbikes (The Unconstrained Utility Benchmark): Trip operations document an unconstrained, highly fluid duty cycle. ICE riders can accept a high-paying, ad-hoc, 30-kilometer delivery dispatch across town at a moment's notice because refueling is instantaneous and universally accessible. Their trip profiles map the natural, uninhibited demand of urban logistics.

2. For EV Motorbikes (The Infrastructure Dependency Map): Trip operations expose the invisible tether of the battery-swapping network. By analyzing individual trip lengths, starting nodes, and destination drop-offs, we can visually diagnose whether EV riders are altering their route choices, avoiding steep topography, or rejecting long-distance fares due to the spatial distribution of swap stations.
Ultimately, transitioning to trip operations allows us to evaluate Asset Productivity per Kilometer. This scale is the single most vital diagnostic tool for determining how vehicle range limits, cargo payloads, and passenger drops impact real-time earning efficiency across the EAC transportation corridor.

```python
trip = pd.read_sql("SELECT * FROM transition_trip",conn)
trip
```

```python
trip.isnull().sum()
```

```python
trip.info()
```

```python
trip.columns
```

```python
import pandas as pd

# 1. Combine and parse string date/time fields into datetime objects
daily_trips = trip.copy()  # Assuming your dataset variable is named 'trip'
daily_trips['start_datetime'] = pd.to_datetime(daily_trips['start_date'] + ' ' + daily_trips['start_time'])
daily_trips['end_datetime'] = pd.to_datetime(daily_trips['end_date'] + ' ' + daily_trips['end_time'])

# 2. Extract operational hour blocks for temporal distribution plotting
daily_trips['start_hour'] = daily_trips['start_datetime'].dt.hour

# 3. Standardize vehicle types to avoid string matching palette dictionary errors
daily_trips['vehicle_type'] = daily_trips['vehicle_type'].str.lower().str.strip()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 5))

# Plot temporal trip density by start hour
sns.kdeplot(
    data=daily_trips,
    x='start_hour',
    hue='vehicle_type',
    palette={'ev': '#2ecc71', 'fuel': '#e74c3c'},
    fill=True,
    alpha=0.3,
    linewidth=2,
    bw_adjust=0.5
)

plt.title('Trip Distribution Across Operational Hour Blocks (Diurnal Patterns)', fontsize=12, fontweight='bold', pad=15)
plt.xlabel('Hour of Day (24h Clock)', fontsize=11, fontweight='bold')
plt.ylabel('Trip Density / Volume', fontsize=11, fontweight='bold')
plt.xticks(range(0, 24))
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

**🔍 Core Analytical Observations from the Output**

1. The Extreme Volumetric Divide (The Scale Disparity):

The diurnal density chart visually reinforces the severe sample size imbalance between the two operating cohorts. The fuel fleet displays massive, towering volume spikes that dominate the graph's area, while the ev cohort flatlines as a thin, barely elevated green band along the absolute bottom margin. This visual profile confirms that EV riders represent an extremely minor, early-stage baseline that fails to register meaningful volume impact across any individual hour block of urban transit demand.

2. The Aggressive Diurnal Pulse Waves (ICE Shift Cascades):

The fuel motorcycle data reveals an incredibly systematic, wave-like operational behavior characterized by deep hourly surges and rapid drops starting from 06:00 and repeating uniformly until 22:00. This highly rhythmic "sawtooth" pattern matches the rapid turnaround cycles of professional Boda-Boda riders.
```python
import matplotlib.pyplot as plt
import seaborn as sns

speed_metrics = ['avg_speed_kmh', 'avg_moving_speed_kmh', 'max_speed_kmh']
colors = {'ev': '#2ecc71', 'fuel': '#e74c3c'}

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, metric in enumerate(speed_metrics):
    sns.kdeplot(
        data=daily_trips,
        x=metric,
        hue='vehicle_type',
        fill=True,
        palette=colors,
        alpha=0.3,
        linewidth=2,
        ax=axes[i]
    )
    axes[i].set_title(f'Velocity Profile: {metric.replace("_", " ").title()}', fontsize=11, fontweight='bold')
    axes[i].set_xlabel(metric.replace('_', ' ').upper())
    axes[i].set_ylabel('Density')
    axes[i].grid(linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Calculate the mean proportion of trip duration lost to idle time per trip
daily_trips['idle_ratio'] = (daily_trips['idle_time_min'] / daily_trips['duration_min']) * 100

avg_idle_ratio = daily_trips.groupby('vehicle_type')['idle_ratio'].mean()

plt.figure(figsize=(8, 6))
bars = plt.bar(avg_idle_ratio.index.map({'ev': 'EV Motorcycle', 'fuel': 'Fuel Motorcycle'}), avg_idle_ratio.values, color=['#2ecc71', '#e74c3c'], width=0.2, alpha=0.85)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.5, f'{yval:.1f}%', ha='center', va='bottom', fontweight='bold')

plt.ylabel('Average % of Trip Duration Spent Idle', fontsize=11, fontweight='bold')
plt.title('Proportion of Non-Productive Idle Time Leaks per Active Trip', fontsize=12, fontweight='bold', pad=15)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

**🔍 Core Analytical Observations from the Output**

1. The Shared Baseline Congestion Toll:

The bar chart highlights a remarkably narrow structural gap in idle time percentages between the two platforms during active transit. EV motorcycles lose an average of 11.4% of their trip duration to idle time, while traditional fuel motorcycles spend 10.4% of their active trip time completely stationary. This highly uniform baseline acts as an empirical signature of urban gridlock confirming that both vehicle types are equally trapped by traffic lights, intersections, and informal street bottlenecks during an active dispatch.

2. The 1% EV Micro-Idling Premium:

While traffic congestion affects both platforms similarly, the EV motorcycle exhibits a distinct 1.0% additional idling penalty. At an active trip level, this small premium suggests subtle operational differences unique to electric vehicles. This can stem from secondary activities like riders pausing briefly mid-trip to double-check battery range metrics on digital dashboards, or minor routing delays as they pass through complex intersections to align with primary energy infrastructure zones.

This 11.4% active-trip idle profile unlocks an incredibly profound structural insight when compared against our earlier Daily Operations diagnostics. In the daily charts, the EV fleet showed massive, long-tailed idle time penalties stretching past 100 minutes. Because this trip chart proves that EV riders are not idling significantly more while out moving passenger or cargo fares, it isolates the problem completely: the severe time leaks crushing EV productivity occur entirely outside active trips specifically during stationary battery swaps, cabinet queues, and charging station turnarounds.

## **Battery End of Life**

Moving to the Battery End-of-Life (EoL) layer brings your diagnostic study to the ultimate closure of the circular economy loop. Up to this point, your value chain analysis has tracked how batteries enter the East African Community (EAC) and how they power localized vehicle deployment. This layer evaluates the inevitable back-end reality: what happens when those battery packs degrade below automotive usefulness (typically at 70–80% State of Health).

Failing to plan for this threshold converts a clean transit movement into an environmental hazard. Lithium-ion batteries contain highly concentrated critical minerals and toxic chemical compounds. Left unmanaged, they represent a significant e-waste burden for municipal landfills. However, when managed through data-driven circular frameworks, they represent a high-value domestic resource:

1. The Second-Life Divergence (Repurposing): Battery packs retired from intensive Boda-Boda or public transit duty cycles are not dead cells. They possess substantial remaining stationary capacity. This data maps the economic and technical viability of harvesting these modules to support secondary storage markets—such as anchoring localized solar mini-grids, backup power for telecom towers, or stabilizing national utility grids (like Kenya Power or TANESCO) during peak hours.

2. The Ultimate Material Recovery (Recycling & Hydrometallurgy): When a battery completes its secondary life and hits absolute degradation, it enters the final extraction phase. This layer tracks the efficiency of domestic hydrometallurgical or pyrometallurgical processing loops. It isolates whether the EAC can successfully recover raw chemical precursors (black mass containing cobalt, nickel, lithium, and manganese) to feed back into regional manufacturing, or if the region will suffer from a material leak by exporting untreated battery scrap.
Ultimately, evaluating the Battery End-of-Life layer allows us to calculate the Residual Value Retention and Lifecycle Carbon Displacement Offset. This is the core indicator for determining if the EAC can establish a self-sustaining, closed-loop green economy that permanently decouples transport from both foreign oil and raw mineral dependencies.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Melt data for clean multi-bar plotting
melted_battery = pd.melt(
    battery, 
    id_vars=['Country'], 
    value_vars=['Mini_Grid_Diversion_Rate', 'E_Waste_Landfill_Diversion_Rate'],
    var_name='Diversion_Type', 
    value_name='Rate'
)

# Aggregate to mean values per country
agg_melted = melted_battery.groupby(['Country', 'Diversion_Type'])['Rate'].mean().reset_index()

# Human-readable labels
agg_melted['Diversion_Type'] = agg_melted['Diversion_Type'].map({
    'Mini_Grid_Diversion_Rate': 'Second-Life Mini-Grid Routing (%)',
    'E_Waste_Landfill_Diversion_Rate': 'Total Landfill Diversion Rate (%)'
})

plt.figure(figsize=(12, 6))
sns.barplot(
    data=agg_melted,
    x='Country',
    y='Rate',
    hue='Diversion_Type',
    palette='viridis',
    alpha=0.85
)

plt.title('Lifecycle Circular Diversion Performance across EAC Countries', fontsize=12, fontweight='bold', pad=15)
plt.xlabel('EAC Member State', fontsize=11, fontweight='bold')
plt.ylabel('Diversion Efficiency Rate (Proportion / %)', fontsize=11, fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# 1. Clean and convert operating hours to numerical daily values if they are strings (e.g., "24 Hours" or "08:00-17:00")
def parse_operating_hours(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).lower().strip()
    if '24' in val_str or 'always' in val_str or 'non-stop' in val_str:
        return 24.0
    # Handle ranges like "08:00-17:00" or "8am-5pm"
    try:
        if '-' in val_str:
            parts = val_str.split('-')
            # Extract basic numeric hour values
            start = float(''.join(c for c in parts[0] if c.isdigit() or c == '.'))
            end = float(''.join(c for c in parts[1] if c.isdigit() or c == '.'))
            # Rough normalization for standard am/pm or 24h formats
            if 'pm' in parts[1] and end < 12: end += 12
            if 'pm' in parts[0] and start < 12: start += 12
            duration = end - start
            return duration if duration > 0 else duration + 24
    except:
        pass
    # Fallback default if numeric parsing fails but valid string exists
    return 9.0

# Create a copy and map numerical operational windows
charging_hours = ev_charge.copy()
if charging_hours['operating_hours'].dtype == 'O':
    charging_hours['daily_hours_open'] = charging_hours['operating_hours'].apply(parse_operating_hours)
else:
    charging_hours['daily_hours_open'] = charging_hours['operating_hours']

# 2. Identify the top 4 infrastructure providers across the whole regional dataset
top_4_providers = charging_hours['provider'].value_counts().nlargest(4).index
filtered_infra = charging_hours[charging_hours['provider'].isin(top_4_providers)]

# 3. Plot the categorical distribution using a clean, easily interpretable factorplot/stripplot
plt.figure(figsize=(13, 7))
sns.stripplot(
    data=filtered_infra,
    x='country',
    y='daily_hours_open',
    hue='provider',
    dodge=True,         # Separates the providers into clear, non-overlapping columns
    jitter=0.2,         # Spreads out identical values for visual clarity
    alpha=0.75,
    size=9,
    palette='Set1'
)

plt.title('Operational Availability Envelopes: Top 4 Providers vs. Country and Daily Operating Hours', fontsize=13, fontweight='bold', pad=15)
plt.xlabel('EAC Member State', fontsize=11, fontweight='bold')
plt.ylabel('Daily Operational Availability Window (Hours Open per Day)', fontsize=11, fontweight='bold')
plt.yticks(range(0, 26, 4))
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(title='Top 4 Infrastructure Operators', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

**🔍 Core Analytical Observations from the Output**

1. The Rigid 24/7 vs. Daytime Binary (Operational Envelopes):

The categorical strip plot reveals that the regional charging infrastructure operates with severe temporal constraints. Strikingly, there are zero stations operating on hybrid middle-tier shifts, creating an infrastructure landscape that leaves late-night operators highly vulnerable to sudden access drops.

## **Readiness Scores**


The Final Synthesis: Readiness Scores & Strategic Alignment

We have reached the analytical culmination of your East African Community (EAC) E-Mobility Diagnostic Study. By synthesizing the micro-level telemetry from rider economics, the localized footprint of charging infrastructure, and the macro-financial modeling of utility revenue potential, we establish the National E-Mobility Readiness Framework.

This final synthesis aggregates multiple tracking dimensions—enabling policies, power grid safety buffers, local industrial value addition, and downstream circular economy loops—into a single Readiness Score per country. This score determines exactly how prepared each member state is to transition into a self-sustaining green industrial transport corridor.

```python
readiness = pd.read_csv('readiness and impact scores.csv')
readiness
```

```python
readiness.columns
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Start with a completely fresh copy of your master readiness dataframe
line_df = readiness.copy()

# Melt the data so Seaborn can cleanly map both score columns as separate colored lines
meltED_df = pd.melt(
    line_df, 
    id_vars=['Country'], 
    value_vars=['Consolidated_Readiness_Score', 'Consolidated_Impact_Score'],
    var_name='Index_Type', 
    value_name='Score'
)

# Human-readable label conversions for presentation clarity
meltED_df['Index_Type'] = meltED_df['Index_Type'].map({
    'Consolidated_Readiness_Score': 'Sovereign Readiness Score (Policy & Infrastructure)',
    'Consolidated_Impact_Score': 'Systemic Impact Score (FX Savings & Utility Revenue)'
})

# 2. Initialize a high-resolution, static plotting canvas
fig, ax = plt.subplots(figsize=(12, 6), dpi=150)
sns.set_theme(style="whitegrid")

# 3. Plot the tracking trajectories using parallel line paths
sns.lineplot(
    data=meltED_df,
    x='Country',
    y='Score',
    hue='Index_Type',
    marker='o',          # Circular anchor nodes for every data point
    markersize=10,
    linewidth=3,
    palette=['#3498db', '#2ecc71'], # High-contrast blue and green profile
    ax=ax
)

# 4. DATA LABELS ENGINE: Place exact numbers right on top of the line markers
for idx, row in meltED_df.iterrows():
    ax.annotate(
        text=f"{row['Score']:.1f}",
        xy=(row['Country'], row['Score']),
        xytext=(0, 10),  # Safe vertical padding offset to prevent text collision
        textcoords="offset points",
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold',
        color='#2c3e50'
    )

# Plain-language labels, ticks, and titles
plt.title('EAC E-Mobility Macro Synthesis: Structural Country Alignment Trends', fontsize=13, fontweight='bold', pad=20)
plt.xlabel('EAC Member State', fontsize=11, fontweight='bold')
plt.ylabel('Synthesized Score (0 to 100 Base)', fontsize=11, fontweight='bold')
plt.ylim(0, 105) # Adds a clean upper safety margin for labels
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.legend(title='Master Diagnostic Indices', loc='lower left', frameon=True)

# 5. EXPORT FILE MANIFEST: Save directly out to a clean graphic
plt.savefig('eac_emobility_readiness_line_trend.png', dpi=300, bbox_inches='tight')
print("✅ Macro synthesis line plot successfully saved as 'eac_emobility_readiness_line_trend.png'!")

plt.show()

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configure cohesive, presentation-ready styling
sns.set_theme(style="whitegrid")
plt.rcParams["font.family"] = "sans-serif"

# Initialize a side-by-side 1x2 subplot layout
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# =====================================================================
# LINE PLOT 1: Absolute Fleet Trajectory (Left Axis)
# =====================================================================
sns.lineplot(
    x="year",
    y="aggregated_ev_stock",
    data=growth_eac,
    color="#27ae60",  # E-mobility green
    linewidth=3.5,
    marker="o",
    markersize=9,
    ax=ax1,
    label="Active EV Fleet Size"
)

# Text labels on the markers to call out specific data points explicitly
for x, y in zip(growth_eac["year"], growth_eac["aggregated_ev_stock"]):
    ax1.annotate(
        f"{int(y):,}",
        (x, y),
        textcoords="offset points",
        xytext=(0, 10),
        ha="center",
        fontsize=9,
        weight="bold",
        color="#1e272c"
    )

ax1.set_title(
    "A: Aggregated EAC Active EV Fleet Scale\n(Longitudinal Stock Expansion 2021-2026)",
    fontsize=12,
    weight="bold",
    pad=12,
)
ax1.set_xlabel("Diagnostic Year", fontsize=11)
ax1.set_ylabel("Total Fleet Volume (Active Units)", fontsize=11, weight="bold")
ax1.set_ylim(0, growth_eac["aggregated_ev_stock"].max() * 1.15)  # Buffer space for annotations

# =====================================================================
# LINE PLOT 2: Velocity Index Metrics - YoY vs CAGR (Right Axis)
# =====================================================================
# Plotting the raw Year-on-Year growth percentages
sns.lineplot(
    x="year",
    y="yoy_growth_percentage",
    data=growth_eac,
    color="#e74c3c",  # Velocity red
    linewidth=3,
    marker="s",
    markersize=8,
    ax=ax2,
    label="YoY Growth Rate (%)"
)

# Overlaying the compounding milestones (converting decimals back into raw percentages)
sns.lineplot(
    x="year",
    y=growth_eac["five_year_cagr_milestone"] * 100,
    data=growth_eac,
    color="#2c3e50",  # Anchoring navy
    linewidth=2.5,
    linestyle="--",
    marker="^",
    markersize=8,
    ax=ax2,
    label="Compounding CAGR Timeline (%)"
)

# Highlight where the policy interventions hit their peak curve acceleration
for x, y in zip(growth_eac["year"], growth_eac["yoy_growth_percentage"]):
    if y > 0:  # Skip baseline year text overlay
        ax2.annotate(
            f"+{y:.1f}%",
            (x, y),
            textcoords="offset points",
            xytext=(0, -18 if x == 2026 else 10),  # Adjust position dynamically to prevent overlap
            ha="center",
            fontsize=9,
            color="#e74c3c",
            weight="bold"
        )

ax2.set_title(
    "B: Fleet Turnover Velocity & Compounding Milestones\n(Comparing Volatile YoY Flows vs. Structural Performance)",
    fontsize=12,
    weight="bold",
    pad=12,
)
ax2.set_xlabel("Diagnostic Year", fontsize=11)
ax2.set_ylabel("Growth Metrics Scale (%)", fontsize=11, weight="bold")
ax2.legend(loc="lower right", frameon=True)

# Format master figure presentation parameters
plt.suptitle(
    "EAC E-Mobility Sector Longitudinal Evolution Dashboard",
    fontsize=15,
    weight="bold",
    y=0.98,
)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set clean, presentation-ready visual parameters
sns.set_theme(style="whitegrid")
plt.rcParams["font.family"] = "sans-serif"

# Initialize a side-by-side 1x2 subplot layout to isolate Stock vs. Velocity
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

# Filter out zero entries to prevent math errors on the logarithmic y-axis
plotted_df = growth_country[growth_country["active_ev_stock"] > 0].copy()

# =====================================================================
# LINE PLOT 1: National Fleet Trajectories (Logarithmic Scale)
# =====================================================================
sns.lineplot(
    x="year",
    y="active_ev_stock",
    hue="country_name",
    style="country_name",
    data=plotted_df,
    linewidth=3,
    markers=True,
    markersize=8,
    palette="Set1",
    ax=ax1,
)

# Apply a base-10 log scale to elegantly handle extreme data disparities
ax1.set_yscale("log")

ax1.set_title(
    "A: National EV Fleet Trajectories \n"
    "Exposing the Multi-Speed Regional Adoption Reality",
    fontsize=13,
    weight="bold",
    pad=12,
)
ax1.set_xlabel("Diagnostic Year", fontsize=11)
ax1.set_ylabel(
    "Active EV Stock Volume", fontsize=11, weight="bold"
)
ax1.legend(title="Partner State", loc="upper left", frameon=True)

# Add clear visual pointers for the main tipping points
ax1.annotate(
    "Kenya: Rapid 2-Wheeler Scale",
    xy=(2024, 12541),
    xytext=(2022, 15000),
    arrowprops=dict(facecolor="#2c3e50", arrowstyle="->", lw=1.5),
    fontsize=9,
    weight="bold",
)

# =====================================================================
# LINE PLOT 2: Year-on-Year Flow Velocity comparison
# =====================================================================
# Isolate the top 4 active transition markets to prevent flat-line clutter on the YoY graph
active_markets = growth_country[
    growth_country["country_iso"].isin(["KEN", "RWA", "UGA", "TZA"])
]

sns.lineplot(
    x="year",
    y="yoy_growth_percentage",
    hue="country_name",
    data=active_markets,
    linewidth=2.5,
    marker="s",
    markersize=7,
    palette="Set1",
    ax=ax2,
)

ax2.set_title(
    "B: Active Market Flow Velocity (YoY %)\n"
    "Tracking Policy Shockwaves & Market Inflections",
    fontsize=13,
    weight="bold",
    pad=12,
)
ax2.set_xlabel("Diagnostic Year", fontsize=11)
ax2.set_ylabel("Year-on-Year Growth Rate (%)", fontsize=11, weight="bold")
ax2.legend(title="Partner State", loc="upper right", frameon=True)

# Format master figure presentation parameters
plt.suptitle(
    "EAC Country-Level Longitudinal E-Mobility Diagnostic Profiles",
    fontsize=16,
    weight="bold",
    y=0.98,
)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Setup the dataframes
valuation_data = {
    'Year': [2025, 2026, 2031, 2025, 2031],
    'Value (USD Billion)': [0.45, 0.69, 5.99, 26.70, 39.90],
    'Segment': ['Core EV Market', 'Core EV Market', 'Core EV Market', 'Broader Ecosystem', 'Broader Ecosystem']
}
df_val = pd.DataFrame(valuation_data)

growth_data = {
    'Segment': ['Mass Transit Bus', '2/3 Wheeler (Micro-mobility)', 'Passenger LDV'],
    'YoY Growth (%)': [44.0, 38.0, 28.0]
}
df_growth = pd.DataFrame(growth_data)

# 2. Create the plots
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
sns.set_theme(style="whitegrid")

# Chart 1: Market Valuations
sns.lineplot(data=df_val, x='Year', y='Value (USD Billion)', hue='Segment', marker='o', ax=axes[0], linewidth=2.5)
axes[0].set_title('Market Valuation Projections (2025–2031)', fontsize=14, pad=10)
axes[0].set_ylabel('Valuation (USD Billion)')

# Chart 2: Segment Growth Rates
sns.barplot(data=df_growth, x='YoY Growth (%)', y='Segment', palette='viridis', ax=axes[1])
axes[1].set_title('Year-over-Year Growth Rate by Segment', fontsize=14, pad=10)
axes[1].set_xlim(0, 50)

plt.tight_layout()
plt.show()
```


🔍 **Core Analytical Observations from the Output**

1. The Ecosystem Value Overwhelmingly Dominates Capital Space: In the market valuation projections, the Broader Ecosystem Value dwarf the Core EV Market Revenue by orders of magnitude. In 2025, the broader auto infrastructure asset ecosystem sits at $26.70 Billion, while the pure-play vehicle sales market revenue registers at just $0.45 Billion. This indicates that the vast majority of capital, value-add, and investment opportunities in the near term lie in heavy supply chain infrastructure, localized assembly networks, charging ecosystems, and auxiliary grid support rather than simple vehicle unit margin distribution.

2. Core EV Market Demonstrates Explosive Compounding Velocity: While starting from a low financial baseline in 2025 ($0.45B), the Core EV Market demonstrates a massive compounding velocity. By leaping past the 2026 inflection point ($0.69B) to arrive at a projected $5.99 Billion valuation by 2031, it captures a 53.90% CAGR. This represents an exponential curve where vehicle revenue begins eating into the broader traditional ecosystem valuation share as scale economies kick in.

3. Mass Transit and Micro-Mobility Outpace Passenger Units in Growth Velocity: The segment growth dynamics clearly isolate Mass Transit Buses (44.0% YoY) and 2/3 Wheelers/Micro-mobility platforms (38.0% YoY) as the primary engines accelerating market expansion. Conversely, premium and consumer-facing Passenger Light-Duty Vehicles (LDVs) lag at 28.0% YoY. This visually proves that the regional growth engine is driven primarily by commercial business-to-business fleet utility rather than private mass-market passenger adoption.

**💡 Strategic Policy Insights for the E-Mobility Diagnosis Study**

1. Pivot Upstream Target Ecosystem Infrastructure Incentives Over Unit Subsidies: Because the broader ecosystem ($26.7B scaling to $39.9B) represents the massive base of the financial pyramid, regional state actors (e.g., ministries of finance or trade) should shift away from consumer-level EV purchase tax rebates. Instead, policies must aggressively target capital deployment incentives for localized assembly infrastructure, battery swapping networks, grid-tied smart charging systems, and raw material supply chains. Securing a fraction of the broader ecosystem yields vastly higher domestic value retention than maximizing pure imported vehicle volumes.

2. De-Risk and Electrify Public Transit Corridors First: With Mass Transit Buses demonstrating the highest underlying momentum (44.0% YoY), municipal and national authorities should focus regulatory muscle on immediate public fleet conversion. This includes mandating electric targets for city bus networks (like the BRT systems in Dar es Salaam or Nairobi) and introducing structured concession agreements for private bus operators. Because public transit vehicles maintain predictable routes and high continuous duty cycles, prioritizing this segment guarantees localized emission reductions and maximizes early charging network utilization.

3. Design Regulatory Standards Custom-Fit for Micro-Mobility Proliferation: Given that 2/3 Wheelers (38.0% YoY) represent the vast volume base across East Africa, regional trade groups must establish unified, standardized protocols for this high-velocity tier. Policy priorities should include unifying battery swapping form factors, implementing universal grid billing systems, and creating formal localized safety certifications for converted light platforms. This ensures that the micro-mobility segment expands into a organized, highly compatible ecosystem rather than splintering into fractured, proprietary battery networks.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up a professional, clean layout
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Plot 1: Cumulative Stock Ranking
# Uses your exact columns: 'Country' and 'Cumulative Stock (2026)'
sns.barplot(
    data=africa_strong.sort_values('Cumulative Stock (2026)', ascending=False),
    x='Cumulative Stock (2026)', 
    y='Country', 
    palette='Blues_r', 
    ax=axes[0]
)
axes[0].set_title('Total EV Fleet Size by Country (2026 Stock)', fontsize=13, weight='bold', pad=12)
axes[0].set_xlabel('Units (Cumulative)')
axes[0].set_ylabel('Country')

# Plot 2: Import Velocity vs Cumulative Stock side-by-side
# Melts your dataset dynamically using your exact column strings
df_melted = africa_strong.melt(
    id_vars=['Country'], 
    value_vars=['Approx Annual Import Count', 'Cumulative Stock (2026)'], 
    var_name='Metric', 
    value_name='Units'
)

sns.barplot(
    data=df_melted, 
    x='Country', 
    y='Units', 
    hue='Metric', 
    palette='GnBu_r', 
    ax=axes[1]
)
axes[1].set_title('Annual Import Velocity vs. Cumulative Stock', fontsize=13, weight='bold', pad=12)
axes[1].set_xlabel('Country')
axes[1].set_ylabel('Number of Vehicles')
axes[1].legend(title='Market Metric')

plt.tight_layout()
plt.show()
```


**🔍 Core Analytical Observations from the Output**

1. Ethiopia's Volume Dominance Rewrites the Continental Ranking: In the 2026 fleet size ranking, Ethiopia stands as the undisputed leader in EV adoption, tracking at a cumulative stock of over 127,000 units. This completely alters the historical narrative of South Africa leading continental clean tech. Ethiopia's stock is more than three times larger than Kenya’s (~37,000 units), six times larger than Egypt’s (~20,000 units), and nearly ten times larger than South Africa’s (~13,500 units).

2. The High-Velocity Inflow Disparity: The side-by-side comparison reveals a massive variance in current market momentum. Ethiopia’s Approx. Annual Import Count (~40,000 units) is exceptionally high relative to its cumulative base. This indicates a highly dynamic, newly accelerated market inflection point. Conversely, South Africa (~4,000 annual imports) and Egypt (~7,000 annual imports) exhibit much flatter ratios, pointing to more conservative, incremental growth paths.

3. Kenya’s Volume Masked by Segment Composition: While Kenya registers a healthy stock of nearly 37,000 units and robust import velocity (~22,000 units annually), this visual parity with industrial giants like Egypt is entirely structural. As previously established, this volume is heavily driven by lightweight micro-mobility platforms (2-wheelers) rather than full-sized passenger or commercial vehicles, meaning its aggregate battery capacity and financial value scale completely differently.

**💡 Strategic Policy Insights for the E-Mobility Diagnosis Study**

1. Leverage Absolute Fiscal Bans for Immediate Structural Transformation: Ethiopia's staggering volume is the direct result of a bold, aggressive policy experiment—the 2024 total ban on internal combustion engine (ICE) passenger vehicle imports. This proves that while minor consumer subsidies or tax exemptions yield slow, linear growth (as seen in South Africa and Egypt), absolute structural restrictions force immediate market conversion. Other African nations seeking rapid decarbonization should study this shock-therapy regulatory model to break vehicle import path dependencies.

2. Proactively Plan for Macro-Grid Infrastructure Bottlenecks: Ethiopia’s explosive addition of ~40,000 electric vehicles annually creates an urgent grid-balancing challenge. While Ethiopia benefits from massive domestic renewable generation (like the Grand Ethiopian Renaissance Dam), its municipal distribution infrastructure must be heavily upgraded. Policy frameworks must immediately mandate smart grid integration, off-peak overnight charging tariffs, and localized storage buffers to prevent localized sub-station overloads in major urban centers.

3. Harmonize Cross-Border Trade Protocols to Unify Fractured Regional Markets: The massive drop-off in cumulative stock between East Africa's frontrunners (Ethiopia and Kenya) and North/Southern Africa highlights regional fragmentation. Trade bodies must use the African Continental Free Trade Area (AfCFTA) framework to unify vehicle safety compliance, zero-emission standards, and tariff exemptions. Creating a friction-free cross-border corridor allows high-volume assembly centers to export vehicles smoothly across regional economic blocs, driving down the unit cost of EVs continent-wide.

```python
glob = pd.read_csv('global_vs_africa_eac_ev_trends.csv')
glob
```

```python
glob['Region / Economic Bloc'].value_counts()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Map names and reshape data as before
region_mapping = {
    'Rest of the World (RoW) (Includes US, China, Europe, Asia)': 'Global',
    'Africa (Total) (Includes Ethiopia & South Africa)': 'Africa',
    'East African Community (EAC) (Excludes Ethiopia)': 'East Africa Community (EAC)'
}
glob['Region / Economic Bloc'] = glob['Region / Economic Bloc'].map(region_mapping).fillna(glob['Region / Economic Bloc'])

year_columns = ['2020 Sales/Footprint', '2022 Sales/Footprint', '2024 Sales/Footprint', '2025 Sales/Footprint', '2026 Projected / YTD']
df_long = glob.melt(id_vars=['Region / Economic Bloc'], value_vars=year_columns, var_name='Year', value_name='Sales_Footprint')
df_long['Year'] = df_long['Year'].str.split(' ').str[0]

# 2. Setup side-by-side subplots (1 row, 2 columns)
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Left Subplot: Global Scale Only
sns.lineplot(
    data=df_long[df_long['Region / Economic Bloc'] == 'Global'],
    x='Year', y='Sales_Footprint', marker='o', linewidth=2.5, color='#4c72b0', ax=axes[0]
)
axes[0].set_title('Global EV Market Trajectory', fontsize=13, weight='bold')
axes[0].set_ylabel('Sales / Footprint Value (Millions)')

# Right Subplot: Africa & EAC Regional Scale Only (Isolated from Global)
regional_data = df_long[df_long['Region / Economic Bloc'].isin(['Africa', 'East Africa Community (EAC)'])]
sns.lineplot(
    data=regional_data,
    x='Year', y='Sales_Footprint', hue='Region / Economic Bloc', 
    marker='o', linewidth=2.5, palette=['#dd8452', '#55a868'], ax=axes[1]
)
axes[1].set_title('Regional African EV Growth Profiles', fontsize=13, weight='bold')
axes[1].set_ylabel('Sales / Footprint Value (Absolute Units)')
axes[1].legend(title='Regional Scope')

plt.suptitle('EV Market Growth Trajectory across Balanced Geographic Scales', fontsize=15, weight='bold', y=1.02)
plt.tight_layout()
plt.show()

```


**🔍 Core Analytical Observations from the Output**

1. Asynchronous Market Inflection Points: The charts reveal a profound timing gap between mature global markets and African ecosystems. The Global trajectory exhibits an intense, continuous compound curve between 2020 and 2024, rising from roughly 3 million to over 17.5 million units before showing a mild deceleration in slope into 2026. Conversely, the African continental market lay entirely dormant until 2022, after which it experienced a sudden, aggressive upward breakout, expanding toward 230,000 units by 2026.

2. The Scale of the EAC Corridor Acceleration: Isolating regional performance exposes the East African Community (EAC) as a primary engine driving sub-Saharan adoption. Standing at approximately 65,000 units in 2026, the EAC alone accounts for nearly 28% of the entire African EV footprint. This confirms that the East African trade corridor is operating as a dense, high-velocity node of alternative powertrain deployment rather than following a passive continental average.

3. Resilience to Global Macro-Flattening: Between 2024 and 2026, the Global plot shows a noticeable rounding and flattening of its line, reflecting localized subsidy rollbacks, trade tensions, and consumer saturation in the EU and US. In stark contrast, the curves for both Africa and the EAC maintain a steep, linear trajectory with zero deceleration, demonstrating that African e-mobility drivers are completely insulated from macro-global cooling patterns.

**💡 Strategic Policy Insights for the E-Mobility Diagnosis Study**

1. Capitalize on Global Overproduction & Asset Liquidation: As the Global adoption curve begins to plateau post-2024, international original equipment manufacturers (OEMs) face a domestic vehicle supply mismatch. Regional economic blocs like the EAC must swiftly enact policies to attract this displaced assembly equipment and component volume. By offering favorable tariff rollbacks on completely knocked-down (CKD) kits rather than fully built imports, local economies can secure rapid technology transfers at historic discount rates.

2. Standardize and Codify the EAC Micro-Mobility Matrix: Given that the EAC commands nearly a third of the continent's volume through distinct, non-global channels (primarily two-wheelers and commercial public transit), its structural rules should serve as the blueprint for neighboring trade blocs (SADC, ECOWAS). Regulators should focus on unifying regional standards for battery swap footprints, continuous multi-hour battery pack testing protocols, and interoperable charging points to prevent a fragmented local market.

3. Insulate Regional Growth via Upstream Continental Integration: While the steep growth profiles in Africa and the EAC prove high domestic market pull, their total reliance on foreign cell manufacturing leaves them exposed to severe supply disruptions. To protect this ongoing upward trajectory through 2030, policymakers must utilize the African Continental Free Trade Area (AfCFTA) framework to forge structural linkages between mineral-rich resource hubs (e.g., lithium/cobalt processing in the DRC) and vehicle assembly hubs across East and Southern Africa, establishing an insulated, self-contained regional value chain.
