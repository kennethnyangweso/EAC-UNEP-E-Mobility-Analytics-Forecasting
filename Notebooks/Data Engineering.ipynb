```python
# Get the existing records from policy_inventory
rows = cursor.execute("""
    SELECT * FROM policy_inventory
""").fetchall()

fixed_data = []

for row in rows:
    text = row[0]

    # Split the first 8 commas only
    parts = text.split(",", 8)

    # First 8 fields are fixed
    first_8 = parts[:8]

    # The remaining part contains description + data_source
    description, data_source = parts[8].rsplit(",", 1)

    fixed_data.append(first_8 + [description, data_source])

# Create corrected DataFrame
columns = [
    "country_iso",
    "year",
    "enacted_policy_instrument",
    "instrument_type",
    "category",
    "status",
    "effective_date",
    "vehicle_segment",
    "description",
    "data_source"
]

df_policy = pd.DataFrame(fixed_data, columns=columns)

print("Shape:", df_policy.shape)
display(df_policy)
```

```python
cursor.execute("""
DROP TABLE IF EXISTS policy_inventory_backup;
""")

cursor.execute("""
CREATE TABLE policy_inventory_backup AS
SELECT * FROM policy_inventory;
""")

conn.commit()

print("Backup created successfully.")
```

```python
df_policy.to_sql(
    "policy_inventory",
    conn,
    if_exists="replace",
    index=False
)

conn.commit()

print("policy_inventory has been corrected and saved.")
```
